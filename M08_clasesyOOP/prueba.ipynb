```python
# Agrupacion de unidades lógicas 
# llamadas OBJETOS
# NOSOTROS como seres humanos vemos en la vida cotidiana
# que los OBJETOS se relacionan entre si, e interactúan entre sí. 
# 
# NUEVO PARADIGMA DE LA PROGRAMACION
# Objetos como base, que contienen información en forma de 
# atributos y código en forma de métodos. 

# Los objetos se crean a partir de una PLANTILLA llamada CLASE
# cada objeto es una INSTANCIA de su Clase

# Ejemplo:
# clase GATO se instancia a Pompón
# Pompón (unn gato en particular) que es un nuevo objeto derivado de la clase GATO

# Sus atributos son: (datos)
# nombre: 'Pompón'
# color: 'Gris'
# raza: 'Siamés' 
# edad: '2 años'
# sexo: 'Macho' 

# Sus métodos son: (funcionalidades)
# Maulla('Miau, Miau')
# Juega
# Come
# Camina

# Los atributos y los métodos se definen al crear la clase GATO
# porque son los atributos y métodos que tienen en común todos los Gatos
# al instanciar la clase GATO que se crea un objeto nuevo llamado 'Pompón' 
# los atributos y métodos de la Clase Padre (plantilla) se HEREDAN al 
# objeto 'Pompón' 

# ************************************
# La programación orientada a objetos (POO) en Python 
# es un paradigma de programación que se basa en el concepto 
# de objetos y clases. 
# En lugar de enfocarse únicamente en las acciones y funciones 
# que se deben realizar, la POO se centra en la creación de objetos, 
# que son instancias de clases. 
# Una clase es una plantilla que define las características 
# y el comportamiento de un objeto.

# Clases y objetos

# Imagina que tienes una plantilla llamada "Clase". 
# Esta plantilla describe cómo se deben crear los objetos. 
# Cada vez que creas un objeto utilizando esa plantilla, 
# se dice que estás "instanciando" la clase.

# Un objeto es como un contenedor que tiene 
# sus propias características y puede realizar acciones específicas. 
# Imagina que la clase es un plano para construir una casa, 
# y cada objeto es una casa construida siguiendo ese plano.

# Los objetos tienen atributos, 
# que son como características o propiedades que los definen. 
# Por ejemplo, en una clase "Persona", 
# los atributos podrían ser el nombre, la edad y la altura de una persona.

# Además de los atributos, los objetos también tienen métodos. 
# Los métodos son como acciones que los objetos pueden realizar. 
# Siguiendo el ejemplo de la clase "Persona", 
# los métodos podrían ser "caminar", "hablar" o "comer". 
# Estos métodos permiten que los objetos realicen acciones específicas.

# Entonces, la programación orientada a objetos en Python 
# se trata de crear objetos a partir de una plantilla llamada clase. 
# Cada objeto es una instancia de su clase, 
# lo que significa que sigue las características 
# y acciones definidas en la clase. 
# Los objetos tienen atributos que los describen y 
# métodos que les permiten realizar acciones.
```

```python
# 1. Declaración de clase: 
# Para definir una clase, utilizamos la palabra clave class 
# seguida del nombre de la clase. 
# Por ejemplo, podríamos tener una clase llamada "Perro".

# 2. Atributos de clase: 
# Los atributos de clase son variables que pertenecen 
# a la clase en su conjunto. 
# Se definen dentro de la clase, 
# pero fuera de cualquier método. 
# Pueden ser accedidos por todas las instancias de la clase. 
# Por ejemplo, un atributo de clase podría ser "especie" con el valor "canino".

# 3. Método __init__: 
# Es un método especial en Python 
# que se llama automáticamente cuando se crea 
# un nuevo objeto de una clase. 
# Se utiliza para inicializar los atributos 
# de instancia del objeto. 
# Se define dentro de la clase y tiene el nombre __init__.

# 4. Atributos de instancia: 
# Los atributos de instancia son variables específicas 
# de cada objeto creado a partir de la clase. 
# Se definen dentro del método __init__ 
# utilizando el parámetro self. 
# Self es 'este' o 'si mismo' y se refiere al futuro del objeto 
# instanciado a traves de la clase
# Por ejemplo, en la clase "Perro", 
# podríamos tener atributos de instancia como "nombre" y "edad".

# 5. Métodos: 
# Los métodos son funciones 
# definidas dentro de la clase 
# y se utilizan para realizar acciones o 
# manipular los atributos de instancia de los objetos. 
# Estos métodos también se definen utilizando 
# el parámetro self para acceder a los atributos y 
# otros métodos de la clase.

# 6. Creación de objetos: 
# Una vez que hemos definido nuestra clase, 
# podemos crear objetos a partir de ella. 
# Para crear un objeto, 
# simplemente llamamos al nombre 
# de la clase como si fuera una función. 
# Por ejemplo, podemos crear un objeto "mi_perro" utilizando la clase "Perro".
```

```python
# class Nombre_clase: # por convención las clases se escriben con Snake_case
#     Constructor de la clase: init 
#     def __init__(self, parametro):

# Ejemplo:
class Gato:
    def __init__(self, nombre, color, raza, edad, sexo, energia):
        # Defino los atributos
        self.nombre = nombre
        self.color = color
        self.raza = raza
        self.edad = edad
        self.sexo = sexo
        self.energia = energia
    
    # Método presentar
    def presentarse(self):
        self.descarga_energia(5)
        return print(f'Nombre: {self.nombre}, Color: {self.color}, Raza: {self.raza}, Edad: {self.edad}, Sexo: {self.sexo}, Energía {self.energia}.')
    
    def maullar(self):
        self.descarga_energia(5)
        return print('Miau Miau Miauuuuuuu')

    def comer(self):
        if self.energia < 100:
            self.carga_energia(10)
            print('El gato está comiendo')
        else:
            print("El gato dejó de comer!")
            self.descarga_energia(5)
    
    def jugar(self):
        if self.energia > 0:
            print('El gato esta jugando...')
            self.descarga_energia(25)
        else:
            print('El gato esta muerto de hambre!')
            
    
    def carga_energia(self, energia):
        self.energia += energia
        
    def descarga_energia(self, energia):
        self.energia -= energia
            
        
```

```python
# Encapsulamiento
# Se utiliza cuando es necesario que ciertos métodos 
# o propiedades sean inviolables o inalterables. 
# Se refiere a la ocultación de los detalles internos 
# de una clase para proteger los datos y comportamientos 
# de esa clase de accesos no autorizados 
# o modificaciones accidentales.

# El encapsulamiento es como un cofre con una cerradura. 
# Los datos y comportamientos de una clase se almacenan 
# dentro del "cofre" y solo se pueden acceder a ellos 
# a través de métodos públicos autorizados, 
# como si tuvieras la "llave" para abrir el cofre. 
# Esto ayuda a prevenir la modificación accidental 
# o no autorizada de los datos y comportamientos de la clase.

class CuentaBancaria:
    
    def __init__(self, saldo): #constructor
        self.saldo_no_encapsulado_o_publico = saldo    # sin encapsular/público
        self.__saldo_encapsulado_o_privado = saldo  # con encapsulamiento/privado
    
    def mostrar_saldo(self):
        print(f"El saldo es {self.saldo_no_encapsulado_o_publico}; (sin encapsular/público)")
        print(f"El saldo es {self.__saldo_encapsulado_o_privado}; (encapsulado/privado)")
        

cta = CuentaBancaria(12.5) #llamando al constructor de la clase
cta.mostrar_saldo()
# modifico el saldo
cta.saldo_no_encapsulado_o_publico = 500.0 #no está encapsulado!!!!
cta.mostrar_saldo()
cta.saldo_encapsualdo_o_privado = 10000 # no accede al atributo de clase #siempre devuelve 12.5
cta.mostrar_saldo()
```

```python
# para poder modificar un dato encapsulado se hace por metodos definidos
# en la clase
class CuentaBancaria:
    def __init__ (self, saldo):
        self.__saldo = saldo 
    
    def saldo_de_cuenta(self):
        print(f"Usted dispone de un saldo de ${self.__saldo}")
    
    def depositar(self, dinero):
        print(f"Usted depositó ${dinero}")
        self.__saldo += dinero
        self.saldo_de_cuenta() 
        
    def retirar(self, dinero):
        print(f"Usted intenta retirar ${dinero}")
        if dinero > self.__saldo:
            print("Saldo insuficiente")
            self.saldo_de_cuenta()
        else:
            print(f"Retiro otorgado")
            self.__saldo -= dinero 
            self.saldo_de_cuenta()
            
cta1 = CuentaBancaria(200.0)
cta1.saldo_de_cuenta()
cta1.retirar(100.5)
cta1.depositar(50.0)
cta1.retirar(5000.0)
```

```python
# correccion por gemini

# La herencia en Python es un concepto fundamental 
# de la Programación Orientada a Objetos (POO) 
# que permite a una clase (llamada clase hija o subclase) 
# heredar los atributos y métodos de otra clase 
# (llamada clase padre o superclase).

# Esto sirve principalmente para reutilizar 
# código y establecer relaciones jerárquicas entre conceptos.

# Ejemplo básico
# Si tienes una clase Animal con un comportamiento general, 
# puedes crear una clase Perro que herede de ella sin necesidad 
# de volver a escribir los mismos atributos y métodos:
```

```python
# Clase Padre (Superclase)
class Animal:

    def __init__(self, nombre):
        self.nombre = nombre

    def comer(self):
        print(f"{self.nombre} está comiendo.")


# Clase Hija (Subclase) -> Se pasa la clase padre entre paréntesis
class Perro(Animal): #Perro hereda de Animal

    def ladrar(self):
        print(f"{self.nombre} dice: ¡Guao!")


# Uso
mi_perro = Perro("Firulais")

# Heredado de la clase Animal
mi_perro.comer()  # Imprime: Firulais está comiendo.

# Propio de la clase Perro
mi_perro.ladrar()  # Imprime: Firulais dice: ¡Guao!

```

```python
# Conceptos clave
# Sobrescritura de métodos (Method Overriding): 
# Una clase hija puede modificar el comportamiento 
# de un método definido en la clase padre.

class Gato(Animal):
    # Sobrescribimos el método comer
    def comer(self):
        print(f"{self.nombre} come despacio y con elegancia.")
        
        
# **La función `super()`:** 
# Permite llamar a métodos de la clase # padre desde la clase hija, 
# especialmente útil para no duplicar 
# la inicialización en `__init__`.

class Ave(Animal):
    #    animal no tiene puede_volar!!! 
       def __init__(self, nombre, puede_volar):
           super().__init__(nombre)  # Inicializa el nombre en la clase padre Animal
           self.puede_volar = puede_volar
           
micho = Gato('Micho')
micho.comer()
# micho.ladrar() tira error!!! (solo atributo nombre y metodo comer [sobreescrito])
# micho.puede_volar  tira error puede volar solo para aves

pajarito = Ave('Pio', True)
pajarito.comer()
pajarito.puede_volar
```

```python
# Herencia múltiple: 
# A diferencia de otros lenguajes de programación, 
# Python permite que una clase hija herede de múltiples clases padre:
    
class Humano:

    def hablar(self):
        print("Hola")


class Robot:

    def cargar_bateria(self):
        print("Cargando...")


# Hereda de ambas clases
class Cyborg(Humano, Robot):
    pass #no hace nada (pass)
            # no se agregan metodos ni atributos

robocop = Cyborg()
robocop.hablar()
robocop.cargar_bateria()

### ¿Por qué utilizar herencia?

# * **Evita la duplicación de código:** 
# Agrupa la lógica común en un solo lugar.
# * **Mantenibilidad:** 
# Si cambias la lógica en la clase padre, 
# automáticamente se actualizan todas las subclases.
# * **Extensibilidad:** 
# Facilita añadir nuevas funcionalidades 
# creando nuevas subclases sin alterar el código existente.
```

```python
# Herrencia
# Para crear una clase especializada, 
# primero defines la clase base que contiene los atributos 
# y métodos generales. 
# Luego, creas una clase hija que hereda de la clase base y 
# agregas nuevos atributos y métodos específicos de la especialización.

# Al heredar de una clase base, 
# la clase hija puede acceder a todos los atributos y 
# métodos de la clase base. 
# Esto te permite utilizar y modificar los atributos y 
# métodos heredados, así como agregar nuevos atributos y métodos propios 
# que son relevantes para la especialización.

# La especialización mediante herencia 
# te permite construir una jerarquía de clases donde 
# las clases hijas heredan características comunes de la clase base 
# y agregan su propia lógica específica. 
# Esto te ayuda a organizar tu código de manera eficiente, 
# evitando la repetición innecesaria de código y 
# promoviendo la modularidad y la reutilización.

# La herencia es compartir comportamientos y atributos de una clase 'Padre' 
# a una clase hijo (superclase a una subclase). 
# La herencia recibida por la subclase se puede reescribir para que la clase
# hija se pueda especializar en un tema concreto 
# (persona (superclase) -> alumno (subclase)).

class Animal: #clase base
    def __init__(self, especie, edad):
        self.__especie = especie
        self.__edad = edad 
    
    # metodo generico pero con implementacion particular 
    # cada clase derivada tendrá su implementacion específica
    def hablar(self):
        # método vacio
        pass 
    
    # metodo generico pero con implementacion particular 
    # cada clase derivada tendrá su implementacion específica
    def moverse(self):
        # método vacio
        pass 
    
    # metodo generico con la misma implementacion particular
    def describime(self):
        print(f'Soy un animal del tipo {type(self).__name__}')
    
```

```python
# ¿Para que vamos a utilizar la herencia?
# Es útil cuando tenemos clases entre sí que se parecen 
# mucho pero cada una tiene cierta particularidad

# Perro hereda de Animal
class Perro(Animal):
    pass 
```

```python
# utilizar super()
# para traer los atributos de la superclase
class Estudiante(Humano):
    def __init__(self, nombre, edad, materia):
        # sin super()
        # self.nombre = nombre
        # self.edad = edad
        
        # con super
        super().__init__(nombre, edad)
        # atributo específico del estudiante
        self.materia = materia
    
    # método específico del estudiante
    def estudiar(self):
        print(f'El alumno {self.nombre} estudia {self.materia} para el examen!')
```
