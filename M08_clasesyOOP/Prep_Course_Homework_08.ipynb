4) Agregar a la clase Vehiculo, un método que muestre su estado, es decir, a que velocidad se encuentra y su dirección. Y otro método que muestre color, tipo y cilindrada

```python
# Vehiculo_M02 subclase de Vehiculo_M01 (los metodos se heredan automaticamente)
class Vehiculo_M02(Vehiculo_M01):
    def __init__(self, color, cilindrada_del_motor, tipo, velocidad, direccion):
        self.velocidad = velocidad 
        self.direccion = direccion
        # el resto de los atributos los traigo de la superclase
        super().__init__(color, cilindrada_del_motor, tipo)
    
    # metodos particulares de esta clase
    def estado_vehiculo(self):
        return f'{self.tipo}: La velocidad es {self.velocidad} km/h y su direccion es {self.direccion}.'

    def presentar_vehiculo(self):
            return f'El tipo del vehículo es {self.tipo}; de color {self.color} con una cilindrada de {self.cilindrada_del_motor} mm3.'
    
```

```python
class Numeros:
    '''Operaciones con números'''
    def __init__(self):
        self.__es_primo = True
        self.__evaluar_nro = 0
        self.__respuesta = tuple()
        self.__moda = list()
        self.__mediana = list()
        self.__media = list()
        self.__valor_grados = 0.0
        self.__valor_destino = 0.0
    
    def cargar_nro_primo(self, numero):
        '''Descripción: Carga y calcula si el número entero ingresado es primo 
        Devolucion: Tupla con formato (numero, True/False) -> Si tal número es primo o no.'''
        self.__evaluar_nro = numero
        self.__es_primo = True
        if isinstance(self.__evaluar_nro, int): 
            # print(f'{self.__evaluar_nro} es entero ({type(self.__evaluar_nro)})')
            if self.__evaluar_nro <= 0:
                print(f'El número cargado -->{self.__evaluar_nro}<-- es igual o menor a cero....')
                return (None, None)
            elif self.__evaluar_nro < 2:
                self.__es_primo = False
            elif self.__evaluar_nro < 4:
                self.__es_primo = True
            # elif self.__evaluar_nro < 2:
                # self.__es_primo = False
            # elif self.__evaluar_nro <= 0:
                # print(f'El número cargado -->{self.__evaluar_nro}<-- es igual o menor a cero....')
                # return (None, None)
            else:
                print(f'Calcular si el numero {self.__evaluar_nro} es primo ....')
                self.__respuesta = self.__calcular_primo()
                self.__evaluar_nro = self.__respuesta[0]
                self.__es_primo = self.__respuesta[1]
                return (self.__evaluar_nro, self.__es_primo)
        else:
            print(f'El número cargado -->{self.__evaluar_nro}<-- no es válido....')
            return (None, None)
        return (self.__evaluar_nro, self.__es_primo)

    def __calcular_primo(self) -> tuple:
        # print(f'Verificar si {self.__evaluar_nro} es un numero primo')
        # self.es_primo = False
        for i in range(2, self.__evaluar_nro):
            if self.__evaluar_nro % i == 0:
                # print('division entera exacta')
                self.__es_primo = False
                break
        return (self.__evaluar_nro, self.__es_primo)
    
    def cargar_nros_estadistica(self, numeros_moda, numeros_mediana, numeros_media):
        self.__moda = numeros_moda
        self.__mediana = numeros_mediana
        self.__media = numeros_media
        print(f'La Moda de {self.__moda} es {self.calcular_la_moda()}.')
        print(f'La Mediana de {self.__mediana} es {self.calcular_la_mediana()}.')
        print(f'La Media de {self.__media} es {self.calcular_la_media()}')

    def calcular_la_moda(self):
        return self.__moda
    
    def calcular_la_media(self):
        return sum(self.__media) / len(self.__media)

    def calcular_la_mediana(self):
        print(self.__mediana.sort())
        if len(self.__mediana) % 2 == 1: #numero impar de elementos
            indice = len(self.__mediana) // 2
            print('El indice nros impar es ', indice)
            return self.__mediana[indice]
        else:
            indice1 = len(self.__mediana) // 2
            indice2 = indice1 + 1
            print('El indice nros impar es ', indice1, 'e indice nros par es ', indice2)
            # print(self.__mediana)
            # print(self.__mediana[indice1 - 1])
            # print(self.__mediana[indice2 - 1])
            return (self.__mediana[indice1 - 1] + self.__mediana[indice2 - 1])/2
        
    def conversion_grados(self, valor, origen, destino):
        self.__valor_grados = valor
        if (origen == 'celsius'):
            if (destino == 'celsius'):
                self.__valor_destino = self.__valor_grados
            elif (destino == 'farenheit'):
                self.__valor_destino = (self.__valor_grados * (9 / 5) + 32)
            elif (destino == 'kelvin'):
                self.__valor_destino = self.__valor_grados + 273.15
            else:
                print('Parámetro de Destino incorrecto')
        elif (origen == 'farenheit'):
            if (destino == 'celsius'):
                self.__valor_destino = (self.__valor_grados - 32) * (5 / 9)
            elif (destino == 'farenheit'):
                self.__valor_destino = self.__valor_grados 
            elif (destino == 'kelvin'):
                self.__valor_destino = ((self.__valor_grados - 32) * (5 / 9) + 273.15)
            else:
                print('Parámetro de Destino incorrecto')
        elif (origen == 'kelvin'):
            if (destino == 'celsius'):
                self.__valor_destino = self.__valor_grados - 273.15
            elif (destino == 'farenheit'):
                self.__valor_destino = ((self.__valor_grados - 273.15) * 9 / 5) + 32
            elif (destino == 'kelvin'):
                self.__valor_destino = self.__valor_grados 
            else:
                print('Parámetro de Destino incorrecto')
        else:   
            print('Parámetro de Origen incorrecto')
        return self.__valor_destino
    
    def factorial(self, numero):
        self.numero = numero
        if self.numero > 1:
            self.numero = self.numero * self.factorial(self.numero - 1)
        return self.numero
    
```

```python
# clase ListaDeNumeros()
class ListaDeNumeros():
    def __init__(self, listado_numeros):
        self.lista = listado_numeros
        self.salida_lista = list()
        self.salida_numero = 0.0
        self.__es_primo = False
        
    def Menu(self):
        print("Menu Principal")
        print("1. De los numeros ingresados calcula cuales son primos")
        print("2. Estadística: Calcular la Moda de los numeros ingresados")
        print("3. Estadística: Calcular la Media de los numeros ingresados")
        print("4. Estadística: Calcular la Mediana de los numeros ingresados")
        print("5. Conversión de Grados: Celsius a Farenheit")
        print("6. Conversión de Grados: Celsius a Kelvin")
        print("7. Conversión de Grados: Farenheit a Celsius")
        print("8. Conversión de Grados: Farenheit a Kelvin")
        print("9. Conversión de Grados: Kelvin a Farenheit")
        print("10. Conversión de Grados: Kelvin a Celsius")
        print("11. Factorial: Calcular el factorial de cada numero")
        print("0. SALIR DEL PROGRAMA")
        opc = int(input("Ingrese su elección: "))
        if opc == 1:
            self.__calcular_primo()
            print(self.salida_lista)
        elif opc == 2:
            repite = self.__calcular_moda()
            print(repite)
        elif opc == 3:
            media = self.__calcular_media()
            print(media)
        elif opc == 4:
            mediana = self.__calcular_mediana()
            # mediana = self.__calcular_mediana()
            print(mediana)
        elif opc == 5:
            print(self.__conversion_grados("celsius","farenheit"))
        elif opc == 6:
            print(self.__conversion_grados("celsius","kelvin"))
        elif opc == 7:
            print(self.__conversion_grados("farenheit","celsius"))
        elif opc == 8:
            print(self.__conversion_grados("farenheit","kelvin"))
        elif opc == 9:
            print(self.__conversion_grados("kelvin","farenheit"))
        elif opc == 10:
            print(self.__conversion_grados("kelvin","celsius"))
        elif opc == 11:
            print(self.__carga_factorial_lista())
        else:
            print("Opción incorrecta (1 - 11)")
    
    def __carga_factorial_lista(self):
        for a in self.lista:
            self.salida_lista.append(self.__factorial_numero(a)) 
        return self.salida_lista
    
    def __factorial_numero(self, nro):
        self.nro = nro
        if self.nro > 1:
            self.nro = self.nro * self.__factorial_numero(self.nro - 1)
        return self.nro
            
    def __conversion_grados(self, origen, destino):
        # convertir la lista de grados a float
        lista_float = list()
        for f in self.lista:
            nro_float = float(f)
            lista_float.append(nro_float)
        
        if (origen == "celsius" and destino == "celsius") or (origen == "kelvin" and destino == "kelvin") or (origen == "farenheit" and destino == "farenheit"):
            return f"La lista en grados {origen} es: {lista_float}"
        elif origen == "celsius" and destino == "farenheit":
            # for grado in lista_float:
            #     conversion_farenheit = (grado * (9 / 5) + 32)
            #     self.salida_lista.append(conversion_farenheit)
            self.salida_lista = [f"{(grado * (9 / 5) + 32):.2f}" for grado in lista_float]
                
        elif origen == "celsius" and destino == "kelvin":
            # for grado in lista_float:
            #     conversion_kelvin = (grado + 273.15)
            #     self.salida_lista.append(conversion_kelvin)
            self.salida_lista = [f"{(grado + 273.15):.2f}" for grado in lista_float]
        elif origen == "farenheit" and destino == "kelvin":
            # for grado in lista_float:
            #     conversion_kelvin = ((grado - 32) * (5 / 9) + 273.15)
            #     self.salida_lista.append(conversion_kelvin)
            self.salida_lista = [f"{((grado - 32) * (5 / 9) + 273.15):.2f}" for grado in lista_float]
        elif origen == "farenheit" and destino == "celsius":
            # for grado in lista_float:
                # conversion_celsius = (grado - 32) * (5 / 9)
                # self.salida_lista.append(conversion_celsius)
            self.salida_lista = [f"{((grado - 32)*(5 / 9)):.2f}" for grado in lista_float]
        elif origen == "kelvin" and destino == "farenheit":
            for grado in lista_float:
                conversion_farenheit = ((grado - 273.15) * (9 / 5)) + 32
                self.salida_lista.append(f"{conversion_farenheit:.2f}")                
        elif origen == "kelvin" and destino == "celsius":
            for grado in lista_float:
                conversion_celsius = (grado - 273.15)
                self.salida_lista.append(f"{conversion_celsius:.2f}")
            
        else:
            print('Parámetros de Origen o Destino incorrectos')

        return f"La conversion en grados {origen} a grados {destino} es:\n{self.salida_lista}"
            
    def __calcular_mediana(self):
        
        # Si solo igualas las variables (lista_copia = self.lista), 
        # ambas apuntarán al mismo bloque de memoria y cualquier cambio afectará a las dos.
        # lista_desordenada = self.lista
        
        # desordenada es copia de lista
        lista_desordenada = self.lista.copy()
        
        # ordenada es copia de lista que la ordena
        lista_ordenada = sorted(self.lista)
        
        print(lista_desordenada)
        print(lista_ordenada)

        # si el numero de elementos es impar
        if len(lista_ordenada) % 2 == 1:
            # indice = len(lista_ordenada) // 2
            return f"El elemento de la mediana es: {lista_ordenada[len(lista_ordenada) // 2]}" 
        # si el numero de elementos es par
        else:
            indice = len(lista_ordenada) // 2
            elemento1 = lista_ordenada[indice - 1]
            elemento2 = lista_ordenada[indice]
            return f"El elemento de la mediana es: {(elemento1 + elemento2)/2}"
    
    def __calcular_media(self):
        self.salida_numero = sum(self.lista) / len(self.lista)
        return f"La media de los numeros de la lista es: {self.salida_numero:.2f}"
    
    def __calcular_moda(self):
        # el dato que mas se repite
        repite = 0                                                              # 0             2 2 4 
        repite_max = 0                                                          # 0             2 2 4
        evaluar_nro_max = 0                                                     # 0             2 2 4 
        for a in self.lista:                                                    # a = 2         3 4 5 6 2
            repite = len([r for i, r in enumerate(self.lista) if r == a ])      # repite = 2    2 4 2 1 
            if repite_max < repite:                                             #true           f t f f
                repite_max = repite                                             #r_max = 2      2 4 4 4
                evaluar_nro_max = a                                             #e_max = 2      2 4 4 4
        return f"El número {evaluar_nro_max} se repite {repite_max} veces como máximo."
    
    def __calcular_primo(self):
        # [2,3,4,5,6]
        for l in self.lista:        # 2
            if l == 2 and l not in self.salida_lista: 
                self.salida_lista.append(l)
            for i in range(2, l):   # 2,2
                if l % i == 0:
                    # print(f"NO SE TIENE EN CUENTA -> l:{l} - i:{i}")
                    self.__es_primo = False 
                    break 
                else:
                    # print(f"l:{l} - i:{i}")
                    self.__es_primo = True 
                    continue
            if self.__es_primo == True:
                # print(f"SE TIENE EN CUENTA -> l:{l}")
                self.__es_primo = False
                if l not in self.salida_lista:
                    self.salida_lista.append(l)
                
                #if l % i == 0:      # false
                    #break
            #    else:
                #self.salida_lista.append(l)
        
```

```python
# crear una clase ayudado de modulos
import math
from scipy.constants import convert_temperature
import statistics
from sympy import isprime
from collections import OrderedDict #elimina los elementos repetidos de una lista

class ListaDeNumerosModulos():
    def __init__(self, listado_numeros):
        self.lista = listado_numeros
        self.salida_lista = list()
        self.salida_numero = 0.0
        
    def Menu(self):
        print("Menu Principal")
        print("1. De los numeros ingresados calcula cuales son primos")
        print("2. Estadística: Calcular la Moda de los numeros ingresados")
        print("3. Estadística: Calcular la Media de los numeros ingresados")
        print("4. Estadística: Calcular la Mediana de los numeros ingresados")
        print("5. Conversión de Grados: Celsius a Farenheit")
        print("6. Conversión de Grados: Celsius a Kelvin")
        print("7. Conversión de Grados: Farenheit a Celsius")
        print("8. Conversión de Grados: Farenheit a Kelvin")
        print("9. Conversión de Grados: Kelvin a Farenheit")
        print("10. Conversión de Grados: Kelvin a Celsius")
        print("11. Factorial: Calcular el factorial de cada numero")
        print("0. SALIR DEL PROGRAMA")
        opc = int(input("Ingrese su elección: "))
        if opc == 1:
            self.__calcular_primo()
            print(self.salida_lista)
        elif opc == 2:
            repite = self.__calcular_moda()
            print(repite)
        elif opc == 3:
            media = self.__calcular_media()
            print(media)
        elif opc == 4:
            mediana = self.__calcular_mediana()
            # mediana = self.__calcular_mediana()
            print(mediana)
        elif opc == 5:
            print(self.__conversion_grados("celsius","farenheit"))
        elif opc == 6:
            print(self.__conversion_grados("celsius","kelvin"))
        elif opc == 7:
            print(self.__conversion_grados("farenheit","celsius"))
        elif opc == 8:
            print(self.__conversion_grados("farenheit","kelvin"))
        elif opc == 9:
            print(self.__conversion_grados("kelvin","farenheit"))
        elif opc == 10:
            print(self.__conversion_grados("kelvin","celsius"))
        elif opc == 11:
            # print(math.factorial(self.lista))
            print(self.__carga_factorial_lista())
        else:
            print("Opción incorrecta (1 - 11)")
    
    def __carga_factorial_lista(self):
        self.salida_lista.clear()
        for a in self.lista:
            self.salida_lista.append(math.factorial(a))
        return self.salida_lista
    #         self.salida_lista.append(self.__factorial_numero(a)) 
    #     return self.salida_lista
    
    # def __factorial_numero(self, nro):
    #     # self.nro = nro
    #     # if self.nro > 1:
    #     #     self.nro = self.nro * self.__factorial_numero(self.nro - 1)
    #     # return self.nro
    #     # return math.factorial(nro)
    #     return math.factorial(self.lista)
            
    def __conversion_grados(self, origen, destino):
        # convertir la lista de grados a float
        lista_float = list()
        for f in self.lista:
            nro_float = float(f)
            lista_float.append(nro_float)
        
        if (origen == "celsius" and destino == "celsius") or (origen == "kelvin" and destino == "kelvin") or (origen == "farenheit" and destino == "farenheit"):
            return f"La lista en grados {origen} es: {lista_float}"
        elif origen == "celsius" and destino == "farenheit":
            # conversion por modulo -->  from scipy.constants import convert_temperature
            self.salida_lista.clear()
            for g in self.lista:
                c_a_f = convert_temperature(g, "C", "F")
                # self.salida_lista.append(c_a_f)
                # evitar que salgan los datos de esta manera: np.float64(xx.x)
                self.salida_lista.append(float(c_a_f))            
        elif origen == "celsius" and destino == "kelvin":
            self.salida_lista = [float(convert_temperature(g,"C","K")) for g in lista_float]
        elif origen == "farenheit" and destino == "kelvin":
            # for grado in lista_float:
            #     conversion_kelvin = ((grado - 32) * (5 / 9) + 273.15)
            #     self.salida_lista.append(conversion_kelvin)
            # self.salida_lista = [f"{((grado - 32) * (5 / 9) + 273.15):.2f}" for grado in lista_float]
            self.salida_lista = [f"{float(convert_temperature(g,"F","K")):.2f}" for g in lista_float]
        elif origen == "farenheit" and destino == "celsius":
            # for grado in lista_float:
                # conversion_celsius = (grado - 32) * (5 / 9)
                # self.salida_lista.append(conversion_celsius)
            self.salida_lista = [f"{((grado - 32)*(5 / 9)):.2f}" for grado in lista_float]
        elif origen == "kelvin" and destino == "farenheit":
            for grado in lista_float:
                conversion_farenheit = ((grado - 273.15) * (9 / 5)) + 32
                self.salida_lista.append(f"{conversion_farenheit:.2f}")                
        elif origen == "kelvin" and destino == "celsius":
            for grado in lista_float:
                conversion_celsius = (grado - 273.15)
                self.salida_lista.append(f"{conversion_celsius:.2f}")
            
        else:
            print('Parámetros de Origen o Destino incorrectos')

        return f"La conversion en grados {origen} a grados {destino} es:\n{self.salida_lista}"
            
    def __calcular_mediana(self):
        # calcular la mediana con un modulo statistics
        mediana = statistics.median(self.lista)
        return f"El elemento de la mediana es: {mediana}"
    
    def __calcular_media(self):
        # calcular la media con modulos
        self.salida_numero = statistics.mean(self.lista)
        return f"La media de los numeros de la lista es: {self.salida_numero:.2f}"
    
    def __calcular_moda(self):
        moda = statistics.mode(self.lista)
        multimoda = statistics.multimode(self.lista)
        repite_max = self.lista.count(moda) # Cuenta cuántas veces está la moda
        return f"El número {moda} se repite {repite_max} veces como máximo.\nTodos los elementos que se repiten son: {multimoda}"
    
    def __calcular_primo(self):
        for p in self.lista:
            res = isprime(p)
            if res == True:
                self.salida_lista.append(p)
        # elimina los elementos repetidos
        self.salida_lista = list(OrderedDict.fromkeys(self.salida_lista))
        
        return f"La lista de numeros primos son: {self.salida_lista}"
```
