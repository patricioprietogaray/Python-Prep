```python
# modificado por ia
from bs4 import BeautifulSoup
import urllib.request

url = "https://es.wikipedia.org/wiki/Ruta_Provincial_11_(Buenos_Aires)"

# 1. Creamos un diccionario con las cabeceras (headers) simulando un navegador
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) '
                  'AppleWebKit/537.36 (KHTML, like Gecko) '
                  'Chrome/120.0.0.0 Safari/537.36'
}

try:
    # 2. Creamos el objeto Request pasando la URL y los headers
    req = urllib.request.Request(url, headers=headers)
    
    # 3. Abrimos la URL usando el objeto Request en lugar de pasar solo el texto
    with urllib.request.urlopen(req) as response:
        html = response.read()
        
    # 4. Procesamos con BeautifulSoup
    soup = BeautifulSoup(html, 'html.parser')
    print("¡Conexión exitosa! Título de la página:")
    print(soup.title.string)

except Exception as e:
    print(f"Ocurrió un error: {e}")
```
