```python
# input interrumpe el flujo de ejecucion hasta que no se culmine 
# el proceso de ingreso de datos.
nume = int(input("Ingrese un número: "))

print("Numero como integer")
print(f"Mostrar el numero tres veces seguidas: {nume*3}")
print(type(nume))
```
