```python
import os
# os.getcwd() -> muestra por consola la ruta actual
directorio_proyecto = os.getcwd()
print(directorio_proyecto)

# entrar al directorio de pruebas
dir_pru = os.path.join(directorio_proyecto, "directorio-prueba")
os.chdir(dir_pru)

# crear una carpeta nueva
if not os.path.isdir("CarpetaNueva"):
    os.makedirs("CarpetaNueva")
print(f"El contenido de la subcarpeta es: {os.listdir()}")

# entrar a la carpeta nueva
dir_nva = os.path.join(dir_pru, "CarpetaNueva")
os.chdir(dir_nva)

# crear un archivo vacio dentro de CarpetaNueva
nombre_archivo = "prueba.txt"

# crear o abrir un archivo
with open(nombre_archivo, 'w') as f: 
    pass 

print(f"El contenido de la subcarpeta es: {os.listdir()}")


# muestro el tamaño de un archivo
print(os.path.getsize(nombre_archivo))


# renombrar el archivo
print(f"Renombrar el archivo de {nombre_archivo} a ArchivoRenombrado.txt")
os.rename(nombre_archivo, "ArchivoRenombrado.txt")
print(f"El contenido de la subcarpeta es: {os.listdir()}")

# elimino el archivo
print(f"Elimino el archivo ArchivoRenombrado.txt")
os.remove("ArchivoRenombrado.txt")
print(f"El contenido de la subcarpeta es: {os.listdir()}")

# subo un nivel de carpeta (a la carp padre)
dir_pad = os.path.dirname(dir_nva)
os.chdir(dir_pad)
print(f"El contenido de la subcarpeta es: {os.listdir()}")

# eliminar una carpeta
os.rmdir(dir_nva)
print(f"El contenido de la subcarpeta es: {os.listdir()}")


# volver a la carpeta del proyecto
os.chdir(directorio_proyecto)



# # renombrar un archivos
# os.rename("CarpetaNueva","CarpetaConNuevoNombre")
# print(f"El contenido de la subcarpeta es: {os.listdir()}")

# # eliminar una carpeta
# os.remove("CarpetaConNuevoNombre")
# print(f"El contenido de la subcarpeta es: {os.listdir()}")
```

```python
# al salir del with el archivo se cierra automaticamente
with open('datos.txt', 'r') as archivo: 
    lineas = archivo.readlines()
    for l in lineas: 
        print(l)
```

```python
# al salir del with el archivo se cierra automaticamente
with open('datos.txt', 'r') as archivo: 
    lineas = archivo.readlines()
    for l in lineas: 
        #reemplazo el \n por un espacio en blanco asi no hay saltos
        print(l.replace('\n',''))
```

```python
# sacarle los saltos de linea al arreglo
with open('datos.txt', 'r') as archivo: 
    contenido = archivo.read()
    lineas = contenido.split('\n')
    print(lineas)
```

```python
#posicion no de linea si no de caracteres recorridos desde le primera 
# linea hasta la ultima
with open('datos.txt', 'r') as archivo: 
    contenido = archivo.read()
    lineas = contenido.split('\n')
    posicion = archivo.tell()
    print(posicion)
```

```python
# iniciar la lectura desde la posicion 7
#posicion no de linea si no de caracteres recorridos desde le primera 
# linea hasta la ultima
with open('datos.txt', 'r') as archivo: 
    archivo.seek(9)
    posicion = archivo.tell()
    print(posicion)
    
    contenido = archivo.read()
    lineas = contenido.split('\n')
    print(lineas)
    
```
