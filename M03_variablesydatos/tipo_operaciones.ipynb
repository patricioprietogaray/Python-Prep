```python
# diferentes tipos de datos
# diferentes tipo de operaciones
# aritméticas, relacionales, asignacion y lógica
```

```python
# operaciones aritmeticas
a=1
b=2
# suma +
print(a+b)
# resta -
print(a-b)
# multiplicacion *
print(a*b)
# potenciacion **
print(a**b)
# division /
print(a/b)
# division muestra el resultado solo en formato entero //
print(a//b)
# division muestra el resto %
print(a%b)
# division por cero da error
try:
    print(a/0)
except ZeroDivisionError:
    print('Macho, si podés dividir por cero sos Dios!')
```

```python
# operadores relacionales
a=1
b=1
c=2
d='1'
print(a==b)
print(b==c)
print(b!=c)
print(a>c)
print(a<=b)
print(b==d)
print(b!=d)
```

```python
# video 9:39
# operadores de asignacion 
# =   -> a = 1  asigno 1 a la variable a

# permite modificar el valor de la variable
# al momento de ejecutar la operacion
# +=   -> a += 2   equivale -> a=a+2

# -=; *=; /=; //=; **=; %=

a=1
print(a)
a+=1
print(a)
a-=2
print(a)
a=10
print(a)
a-=5
print(a)
```

```python
# Operar con strings
# concatenacion -> "hola "+"Pepe"
# multipliacacion -> "Ja "*3 -> "Ja Ja Ja "
saludar='Hola '
nombre='Pedro'
exclamacion='!'
saludo_impreso = saludar+nombre+(exclamacion*3)
print(saludo_impreso)
```

```python
# Operaciones Lógicas
# Utilizando variables booleanas (bool)
# Utilizar tablas de verdad and, or, not
```

```python
# si las condiciones se cumplen (true) entonces es True
print("Operador lógico AND")
print(False and False) # False
print(True and False) # False
print(False and True) # False
print(True and True) # True

a=0 #False
b=0
print(a and b) #False (0)
a=1 #True
b=0
print(a and b) #False (0)
a=0
b=1
print(a and b) #False (0)
a=1
b=1
print(a and b) #True (1)
```

```python
# Si al menos una condicion se cumple (True) entonces es True
print("Operador lógico OR")
print(False or False) # False
print(True or False) # True
print(False or True) # True
print(True or True) # True

a=0 #False
b=0
print(a or b) #False (0)
a=1 #True
b=0
print(a or b) #True (1)
a=0
b=1
print(a or b) #True (1)
a=1
b=1
print(a or b) #True (1)
```

```python
# Niega el valor (si es True y lo niego entonces es False)
# si es False y lo niego entonces es True
print("Operador NOT") 
print(not True)
print(not False)
a=1
b=0
print(not a) #False (0)
print(not b) #True (1)
```

```python
# or exclusiva o xor ^
# es true si un operador es verdad y falso el resto
print("Operador lógico XOR")
print(False ^ False) # False
print(True ^ False) # True
print(False ^ True) # True
print(True ^ True) # False

a=0 #False
b=0
print(a ^ b) #False (0)
a=1 #True
b=0
print(a ^ b) #True (1)
a=0
b=1
print(a ^ b) #True (1)
a=1
b=1
print(a ^ b) #False (0)
```
