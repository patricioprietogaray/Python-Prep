18) Utilizando como operandos el número 4 y los resultados obtenidos en los puntos 16 y 17. Obtener 27 como resultado

```python
num_cuat = 4
print(f"Numero Cuatro:  {num_cuat:08b}")  # Muestra: 00000100
num_res_16 = 6
print(f"Numero Seis:    {num_res_16:08b}")  # Muestra: 00000110
num_res_3 = 3
print(f"Numero Tres:    {num_res_3:08b}")  # Muestra: 00000011
print('-'*24)
print(f"Numero 27:      {27:08b}")  # Muestra: 00011011         16+8+2+1 = 27
# 4 and 6  al resultado desp 2 bits a la izq y ese resul and 3
res = num_cuat | num_res_16
print(f"Resultado cua or (|) 16:    {(num_cuat | num_res_16):08b}")  # Muestra: 00000110
# desplazo 2 bits a la izquierda quedando 00011000
res_desp = res << 2
print(f"Resultado desp 2 a la izq:    {res_desp:08b}")  # Muestra: 00011000
res_final = res_desp | num_res_3
print(f"Resultado final (res_desp or num_res_3):    {res_final:08b}")  # Muestra: 00011000
print(f"Equivalente {res_final}")
```
