```python
x=5 # variable de tipo de dato entero
# x es el identificador (como lo voy a llamar cuando
# necesite su dato (5))
# Enteros (int): números enteros positivos o negativos, 
# sin decimales.
# Flotantes (float): números con decimales.
# Booleanos (bool): valores que representan verdadero o falso.
# Cadenas de caracteres (str): secuencias de caracteres o texto.
```

```python
print(x) # muestro por pantalla el valor de x
```

```python
y=2.3 # variable de tipo de dato float (decimal)
print(y) # muestro por pantalla el valor de y
```

```python
z="Hola" # variable de tipo de dato string (cadena de caracteres)
print(z) # print es una funcion y sirve para recuperar o mostrar 
            # el dato de una variable/constante guardada
z_1='Hola'
print(z_1)
```

```python
# type permite mostrar el tipo de dato que contiene una
# variable / constante.
print(type(x))
print(type(y))
print(type(z))
```

```python
# funcion bool -> true / false
# una variable es true si tiene valor almacenado
# una variable es false si esta vacia o es nula 
print(bool(x))
num=0
print(bool(num))
num=None
print(bool(num))
```
