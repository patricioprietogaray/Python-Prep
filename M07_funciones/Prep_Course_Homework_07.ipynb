1) Crear una función que reciba un número como parámetro y devuelva True si es primo y False si no lo es

```python
def es_numero_primo(numero):
    '''Calcula si el numero es primo o no'''
    es_primo = True
    if numero < 2:
        es_primo = False
    elif numero == 2:
        es_primo = True
    else:
        for divisor in range(2, numero):
            # print(f"Evaluando si {numero} es divisible por {divisor}")
            if numero % divisor == 0:
                es_primo = False
                break
    return es_primo
```

```python
numero_a_verificar = 6

primo = es_numero_primo(numero_a_verificar)

if primo:
    print(f'El numero {str(numero_a_verificar)} es primo?: {primo}')
else:
    print(f'El numero {str(numero_a_verificar)} es primo?: {primo}')
```

3) Crear una función que al recibir una lista de números, devuelva el que más se repite y cuántas veces lo hace. Si hay más de un "más repetido", que devuelva cualquiera

```python
def repeticion(lista):
    '''Calcula cuales son los numeros que se repiten en una lista'''
    lista_repetidos = []
    for i in lista:
        if lista.count(i) > 1 and i not in [x[0] for x in lista_repetidos]:
            lista_repetidos.append((i,lista.count(i)))
    # return lista_repetidos
    # el que mas se repite
    for n in lista_repetidos:
        if n[1] == max([x[1] for x in lista_repetidos]):
            return n

lista = [1, 2, 3, 4, 5, 2, 1, 2, 3, 2]
listado = repeticion(lista = lista)
print(listado)

```

4) Crear una función que convierta entre grados Celsius, Farenheit y Kelvin<br>
Fórmula 1	: (°C × 9/5) + 32 = °F<br>
Fórmula 2	: °C + 273.15 = °K<br>
Debe recibir 3 parámetros: el valor, la medida de orígen y la medida de destino


```python
def conversion(valor,origen,destino):
    '''Convierte un valor de una unidad a otra'''
    if origen == 'c'.upper() and destino == 'f'.upper():
        #convertir de C a F
        return (float(valor) * (9/5)) + 32
    elif origen == 'c'.upper() and destino == 'k'.upper():
        #convertir de C a K
        return float(valor) + 273.15
    elif origen == 'f'.upper() and destino == 'c'.upper():
        #convertir de F a C
        return ((float(valor) - 32) * (5/9))
    elif origen == 'f'.upper() and destino == 'k'.upper():
        #convertir de F a K
        return ((float(valor) - 32) * (5/9)) + 273.15
    elif origen == 'k'.upper() and destino == 'c'.upper():
        #convertir de K a C
        return float(valor) - 273.15
    elif origen == 'k'.upper() and destino == 'f'.upper():
        #convertir de K a F
        return ((float(valor) - 273.15)*(9/5)) + 32


```

6) Armar una función que devuelva el factorial de un número. Tener en cuenta que el usuario puede equivocarse y enviar de parámetro un número no entero o negativo

```python
def factorial(numero):
    if numero > 0:
        return numero * factorial(numero - 1)
    else:
        return 1

print("Funcion Factorial!")
while True:
    entrada = input("Ingrese un número positivo mayor que uno para factorizarlo: ")
    try:
        numero_positivo = int(entrada)
        if numero_positivo > 1:
            # print("Es un numero entero positivo!")
            resultado = factorial(numero_positivo)
            break 
        else:
            print("El número debe ser mayor a uno")
            continue 
    except ValueError:
        print("Debe ingresar un nḿero entero positivo sin letras y sin decimales.")
print(f"El resultado del factorial de {numero_positivo}! es {resultado}")
    
```
