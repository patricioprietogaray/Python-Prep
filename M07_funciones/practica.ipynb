```python
# las funciones se declaran y se define su estructura
# las funciones se ejecutan (ejecutar el codigo definido)

# cada vez que la funcion se ejecuta se genera un contenedor
# donde las variables de dicha funcion van a vivir,
# una vez que se sale del contenedor las variables dejan de existir
# contenedor -> scope o espacio de nombres

# las variables que se declaren dentro de la funcion -> variables LOCALES
# las variables que se declaren fuera de la funcion -> variables GLOBALES

# las variables globales pueden ser accedidas dentro de funciones y clases
# las variables declaradas en una funcion no pueden se accedidas desde otro 
# lugar que no sea la funcion declarada
```

```python
def dividir(dividendo, divisor=1):
    if divisor == 0:
        return 'No se puede dividir por cero!'
    else:
        resultado = dividendo // divisor
        resto = dividendo % divisor
    return resultado, resto
```

```python
# ¿Cómo se estructura?

# Para definir una función en Python se utiliza la palabra reservada def, 
# a continuación viene el nombre de la función que es el que se utilizará 
# luego para invocarla. 
# Después del nombre hay que incluir los paréntesis y 
# dentro de ellos una lista opcional de parámetros. 
# Finalmente colocamos  dos puntos. 
# Luego de  los dos puntos se incluye el cuerpo de la función, 
# con una sangría.En el cuerpo de la función van el conjunto de 
# instrucciones que se encapsulan en dicha función.  
# En último lugar y de manera opcional, se añade la instrucción 
# con la palabra reservada return para devolver un resultado.

# Los nombres no pueden comenzar con dígitos
# No se puede utilizar una palabra reservada
# Las variables deben tener diferentes nombres
# Los nombres de las funciones deben ser descriptivas de lo que hacen las funciones
# Los parámetros pueden tener valores por defecto
# Una función puede devolver ninguno, uno o más de un valor, y de diferentes tipos de datos

# ¿Dónde acceder a las variables?

# Cada vez que una función se ejecuta se genera un contenedor, 
# donde las variables de la función van a vivir, 
# una vez se sale de la función estas variables dejan de existir, 
# este contenedor se denomina espacio de nombres, 
# alcance o scope de la función. Dentro de una función puede 
# haber una variable que solo se pueda utilizar dentro 
# del cuerpo de la función, estas son las  variables locales. 
# Mientras que las variables en el código desde el cual se 
# llama a la función, son variables globales.

# Una variable declarada fuera de cualquier función o clase tiene un alcance global. Esto significa que puede ser accedida desde cualquier lugar del programa, incluyendo dentro de funciones y clases.
# Una variable declarada dentro de una función o clase tiene un alcance local. Esto significa que solo puede ser accedida dentro de esa función o clase, y no fuera de ella.









```

```python
# RECURSIVIDAD

# Una función recursiva es una función que se llama a sí misma.
```

```python
# Factorial de un numero n es el producto de todos los enteros positivos menores o iguales a n. 
# Se denota como n! y se define de la siguiente manera:
# n! = n * (n-1) * (n-2) * ...
```

```python
def factorial(numero):
    # numero positivo entero mayor a 1
    if numero > 1:
        numero = numero * factorial(numero - 1)
    return numero
```

```python
factorial_5 = factorial(5)
print(f"El factorial de 5 es: {factorial_5}")
```

```python
# documentar una funcion con docstring ("""
# Esta es la documentacion de la funcion
# """)

def factorial(numero):
    """
    Calcula el factorial de un número entero positivo.
    
    Args:
        numero (int): Un número entero positivo.
        
    Returns:
        int: El factorial del número ingresado.
    """
    # numero positivo entero mayor a 1
    if numero > 1:
        numero = numero * factorial(numero - 1)
    return numero
```
