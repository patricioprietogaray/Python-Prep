# loc e iloc

loc e iloc son dos modos de indexar y cortar datos en un DataFrame o una Serie.
loc en base -> nombre indice / columnas; tipo de slicing inclusivo en ambos extremos. 
iloc -> posiciones enteras: tipo de slicing exclusivo en el extremo final (como python).


```python
# recorrerlo en base a las etiquetas
import pandas as pd 
todo = pd.Series(['Matemáticas','Fisica','Química'], index=['Adrian','Fernanda','Jorge'])
todo
```

```python
todo.loc['Fernanda']
```

```python
todo.iloc[2]
```

Vamos a lo grande y vamos a temporizar las operaciones

```python
import pandas as pd 
import numpy as np 
# cargar 1000000 de numeros aleatorios del 0 al 1000
numbers = pd.Series(np.random.randint(0,1000,1000000))
# mostrar los primeros 5 valores
numbers.head()
```

```python
%%timeit -n 100 #temporiza el algoritmo (random numero bajo)
total = 0
for i in numbers:  
    total += i
total/len(numbers)
# resultado: 2 ms ± 216 μs per loop (mean ± std. dev. of 7 runs, 100 loops each)
# 
# resultado de 1millon y 2m 6.0s: 
#   180 ms ± 5.84 ms per loop (mean ± std. dev. of 7 runs, 100 loops each)
```

```python
%%timeit -n 100 #
total = np.sum(numbers)
total/len(numbers)
# resultado: 72.2 μs ± 23.8 μs per loop (mean ± std. dev. of 7 runs, 100 loops each)

# resultado de 1 millon -> tienpo 5.0 seg
    # 811 μs ± 180 μs per loop (mean ± std. dev. of 7 runs, 100 loops each)
```

Dataframes

La estructura estrella de Pandas, es una tabla en 2-D en donde:
* cada fila es una observación (registro)
* cada columna es una variable (campo / atributo)

# Borrar de manera permanente el indice 2

crear una copia de seguridad del dataframe antes de modificarlo de manera permanente!

```python
import pandas as pd 

datos = {
    'Nombre': ['Pepe', None,'Marcelo', 'Marta'],
    'Producto': ['Mouse', 'Teclado', 'Notebook', 'Parlante'],
    'Categoria': ['Accesorios','Accesorios','Articulo','Sonido'],
    'Precio': [8.00, 6.50, 120.5, 14.8],
    'Cantidad': [1,3,2,1],
    'Fecha': ['2024-6-1','2016-1-19','2022-12-18', '2026-11-6'],
    'Ciudad': ['Lima','Asuncion','Bogota', 'Buenos Aires'],
    'Satisfaccion': [7,6,9,10]
}

df = pd.DataFrame(datos)



# mostrar el original
df

# copio el dataframe original a una copia destino (df_copia)
df_copia = df.copy()

# borro de manera permanete el indice 2 con inplace
# df_copia.drop(2, inplace=True) -> forma antigua
df_copia = df_copia.drop(2) #-> forma moderna

# mostrar el contenido de df_copia
df_copia
```

```python
import pandas as pd 

datos = {
    'Nombre': ['Pepe', None,'Marcelo', 'Marta'],
    'Producto': ['Mouse', 'Teclado', 'Notebook', 'Parlante'],
    'Categoria': ['Accesorios','Accesorios','Articulo','Sonido'],
    'Precio': [8.00, 6.50, 120.5, 14.8],
    'Cantidad': [1,3,2,1],
    'Fecha': ['2024-6-1','2016-1-19','2022-12-18', '2026-11-6'],
    'Ciudad': ['Lima','Asuncion','Bogota', 'Buenos Aires'],
    'Satisfaccion': [7,6,9,10]
}

df = pd.DataFrame(datos)



# mostrar el original
df

# copio el dataframe original a una copia destino (df_copia)
df_copia = df.copy()

# borro de manera permanete el indice 2 con inplace
# df_copia.drop(2, inplace=True) -> forma antigua
df_copia = df_copia.drop('Cantidad', axis=1) #-> forma moderna
# axis=1 -> columna  / axis=0 -> filas

# mostrar el contenido de df_copia
df_copia
```

```python
data_csv = pd.read_csv('machine-readable-business-employment-data-june-2026-quarter.csv')
data_csv.head()
```
