Array

una dimension

```python
# inicializa un arreglo de tres elementos
a = np.array(['a','b','c'])
print(a) # -> ['a' 'b' 'c']
print(a.ndim) # -> 1 muestra el total de dimensiones.
```

dos dimensiones

```python
b = np.array(
    [
        [1,2,3],   #dimension linea (horizontal 1 fila)
        [4,5,6]    #dimension linea (horizontal 1 fila)
        #dimension columna (vertical 3 columnas)
    ]
)   
# dos dimensiones en total
# linea 1 -> 1,2,3 y linea 2 -> 4,5,6
# columna 1 -> 1,4   columna 2 -> 2,5   y columna 3 -> 3,6
print(b)
print(b.ndim)
```

# Indexing y Slicing

indexing    -> arreglo[indice]

```python
print(c)
print(c[1]) # -> np.int64(2)
print(c[-1]) # -> np.int64(9)

# indice -> 0 1 2 3 4 5 6 7 8 
# dato ->  [1 2 3 4 5 6 7 8 9]
```

posicion en arrays de dos dimensiones

```python
d = np.array([[1,2,3],[4,5,6]])
d
```

```python
d.dtype
```

```python
# visualizar la estructura del arrego
d.shape # (2 filas, 3 columnas)
```

```python
# array([[1, 2, 3],
#        [4, 5, 6]])
print(f"cuatro: {d[1,0]}")
# arreglo[fila, columna] -> comienzan por cero
#           y      x

print(d[0,1]) # ->2
print(d[1,2]) # ->6
```

varios datos de un mismo arreglo

```python
#  [1,2,3],
#  [4,5,6]
d = np.array([[1,2,3],[4,5,6]])
# varios datos de un mismo arreglo
print(f"{d[0,0],d[1,2]}")
# fil,col   |      |
#  esultado 1      6
```

# Slicing (obtener un arreglo secundario de un arreglo primario)

```python
a
```

```python
a[1:]   #   posicion 0 1 2
        #   valor    1 2 3
    # arreglo[posicionInicial:posicionFinal]
    # si inicial no esta vacio comienzo desde el valor del indice 
    # y muestro el resto de los elementos del arreglo si esta vacio
    # comienza de la posicion 0
    # a[1:]  comienza desde el indice 1 hasta el final 
    # a[:3]  comienza desde el indice 0 hasta el final en 3
```

```python
a[0:]
# incluye el cero hasta el final   [1,2,3]
```

```python
a[:2]
# excluye el indice dos  [1,2]
```

```python
a[:]  #recorrer todo el arreglo
```

```python
#  array([1, 2, 3])
a[:10] # excluye la posicion 10 -> 0 a 9
# las posiciones que no estan simplemente no las muestra
# no genera error
```
