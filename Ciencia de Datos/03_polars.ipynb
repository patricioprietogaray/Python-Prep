# Comparar Pandas y Polars

```python
# librerias de datos
import pandas as pd 
import polars as pl 

# libreria para controlar el tiempo
import time

# controlar que hora es y guardar el dato
start = time.time()

# leer un archivo parquet
df_lazy_mode = pl.scan_parquet('0000.parquet')

# calcular el tiempo transcurrido en la carga del archivo parquet
load_time_lazy = time.time() - start

print(f"El tiempo transcurrido en la lectura y darse cuenta que hay un archivo partquet con POLARS es {load_time_lazy} segundos")


start = time.time()

# cargar el archivo
df_pl = df_lazy_mode.collect()

load_time_pl = time.time() - start
print(f"El tiempo transcurrido en la carga completa del archivo partquet con POLARS es {load_time_pl} segundos")


# # Archivo con pandas
start = time.time()
df_pandas_mode = pd.read_parquet('0000.parquet')
load_time_pandas = time.time() - start
print(f"El tiempo transcurrido en la carga del archivo partquet con PANDAS es {load_time_pandas} segundos")
```
