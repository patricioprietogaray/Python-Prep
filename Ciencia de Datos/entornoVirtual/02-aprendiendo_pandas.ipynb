Método describe(): Estadística

```python
print(f"\nDescribe: \n{df_filas.describe()}")
# se utiliza en series o dataframes, devuelve un dataframe.
# Count: Número de observaciones no nulas
# Mean: calcula el promedio aritmético de los valores numéricos 
#       a lo largo de un eje especificado (columnas o filas)
# Std: Desviación estandar de los valores.  
#       Indica qué tan dispersos o alejados están los valores de una 
#       columna respecto a su valor promedio (mean). 
#       Un valor bajo de std significa que los datos están muy agrupados 
#       cerca de la media; un valor alto indica que están muy dispersos
# Min: Valor mínimo
# 25%: Primer Cuartil (Q1): Divide el 25% inferior de la distribución del resto de los datos.
# 50%: representa la mediana del conjunto de datos de esa columna
# 75%: representa el tercer cuartil (Q₃) de la distribución de los datos numéricos
# Max: Valor máximo
# **** otros datos 
# Suma: Suma de todos los valores
print(f"\nSum: \n{df_filas.sum()}")
# Abs: Valor absoluto, devuelve un nuevo objeto 
#       (Series o DataFrame) que contiene el valor 
#       absoluto numérico de cada elemento.
print(f"\nValor absoluto:\n{df_filas.abs()}")
# print(f'\nProducto del eje 0 con desbordamiento:\n{df_filas.prod(axis=0)}')
print(f'\nProducto del eje 0 sin desbordamiento:\n{df_filas['Ventas'].astype('float64').prod()}')
# falta cumsum y cumprod
print(f'\nSuma acumulada:\n{df_filas['Ventas'].astype('int64').cumsum()}')
print(f'\nProducto acumulativo:\n{df_filas['Ventas'].astype('float64').cumprod()}')
```

Método Where (Buscar - Buscar/Reemplazar)

```python
import pandas as pd 
import numpy as np
s = pd.Series(np.arange(0,10))
s
```

```python
# filtrar todos los numeros pares de s, 
# NaN para los que no cumplen con la condicion
s.where(s % 2 == 0)
```

```python
# reemplazamos el valor que no cumple por -1
s.where(s % 2 == 0, -1)
```

```python
# ss = s.drop(s.where(s % 2 != 0)) # devuelve valores no indices
# todos los s que cumplan con la condicion s%2==0
# como el mark 

df_quedan = s[s%2==0]
print(df_quedan)
# eliminar por indice con .index
ss = s.drop(s[s%2==0].index)
print(ss)
```

```python
import pandas as pd 
import numpy as np
df = pd.DataFrame(
    np.arange(18).reshape([6,3]),
    index = ["a","b","c","d","e","f"],
    columns = ["A","B","C"]
)
print(df)
# elimino toda la fila
df = df.drop(['a']) #axis=0 -> filas
print(df)
# elimino toda la columna A
df = df.drop(['A'], axis=1)
print(df)
```

```python
import pandas as pd
a = pd.Series(
    [1,2,3,4,5],
    index=["a","b","c","d","e"]
)
b = pd.Series(
    [10,11,12],
    index=["a","b","f"]
)
pd.concat([a,b],axis=0)
# a     1
# b     2
# c     3
# d     4
# e     5
# a    10
# b    11
# f    12
pd.concat([a,b],axis=1)

#   0	1
# a	1.0	10.0
# b	2.0	11.0
# c	3.0	NaN
# d	4.0	NaN
# e	5.0	NaN
# f	NaN	12.0

pd.concat([a,b], axis=1, sort=True) # ????

#   0	1
# a	1.0	10.0
# b	2.0	11.0
# c	3.0	NaN
# d	4.0	NaN
# e	5.0	NaN
# f	NaN	12.0
```

```python
import numpy as np
import pandas as pd 

df1 = pd.DataFrame(np.arange(9).reshape([3,3]), index=['a','b','d'], columns=['A','B','C'])
print(df1)
df2 = pd.DataFrame(np.arange(12).reshape([4,3]), index=["a","b","c","e"], columns=['B','C','D'])
print(df2)
pd.concat([df1,df2], sort=False)
```

```python
# por defecto join="outer" muestra 
# todas las filas 
# aún las que tienen NaN 
pd.concat([df1,df2], sort=False, axis=1)
```

```python
# join inner muestra solo las filas que no contengan NaN
pd.concat([df1,df2], sort=False, axis=1, join="inner")
```
