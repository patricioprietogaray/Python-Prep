Método describe(): Estadística

```python
print(f"\nDescribe: \n{df_filas.describe()}")
# se utiliza en series o dataframes, devuelve un dataframe.
# Count: Número de observaciones no nulas
# Mean: calcula el promedio aritmético de los valores numéricos 
#       a lo largo de un eje especificado (columnas o filas)
# Std: Desviación estandar de los valores.  
#       Indica qué tan dispersos o alejados están los valores de una 
#       columna respecto a su valor promedio (mean). 
#       Un valor bajo de std significa que los datos están muy agrupados 
#       cerca de la media; un valor alto indica que están muy dispersos
# Min: Valor mínimo
# 25%: Primer Cuartil (Q1): Divide el 25% inferior de la distribución del resto de los datos.
# 50%: representa la mediana del conjunto de datos de esa columna
# 75%: representa el tercer cuartil (Q₃) de la distribución de los datos numéricos
# Max: Valor máximo
# **** otros datos 
# Suma: Suma de todos los valores
print(f"\nSum: \n{df_filas.sum()}")
# Abs: Valor absoluto, devuelve un nuevo objeto 
#       (Series o DataFrame) que contiene el valor 
#       absoluto numérico de cada elemento.
print(f"\nValor absoluto:\n{df_filas.abs()}")
# print(f'\nProducto del eje 0 con desbordamiento:\n{df_filas.prod(axis=0)}')
print(f'\nProducto del eje 0 sin desbordamiento:\n{df_filas['Ventas'].astype('float64').prod()}')
# falta cumsum y cumprod
print(f'\nSuma acumulada:\n{df_filas['Ventas'].astype('int64').cumsum()}')
print(f'\nProducto acumulativo:\n{df_filas['Ventas'].astype('float64').cumprod()}')
```

DataFrames -> Mientras que Serie es un vector; dataframe es una matriz (formato diccionario)

```python
# Create
import pandas as pd 
ventas = pd.DataFrame({
    "Entradas":[41,32,56,18],
    "Salidas":[17,54,6,78],
    "Valoracion":[66,54,49,66],
    "Limite":["No","Si","No","No"],
    "Cambio":[1.43,1.16,-0.67,0.77]},
    index=["Ene","Feb","Mar","Abr"])
print(ventas)
```

```python
# Read leer el mes de Feb completo
print(ventas.loc['Feb'])
```

```python
# agregar una columna o atributo 
# con el valor 0.10 a todos las filas del dataframe
ventas['Descuento'] = 0.10
print(ventas)
```

```python
# agregar una nueva fila
# df.loc[indice] = [valor1,valor2,...]
ventas.loc['May'] = [99,99,99,'Si',-0.99, 0.00]
print(ventas)
```

```python
# concateno un df con otro dataframe
# 1. Pasamos el índice explícitamente como una lista: index=[mes]
mes = 'Dic'
nueva_fila = pd.DataFrame([{
            'Entradas': 199, 
            'Salidas':100, 
            'Valoracion': 55, 
            'Limite':'Si', 
            'Cambio':-1.99, 
            'Descuento': 0.50
        }], index=[mes])
# 2. Concatenamos SIN el parámetro ignore_index=True
ventas = pd.concat([ventas, nueva_fila])
print(ventas)
# con ignore_index -> de esta forma elimina los indices (meses) y los reemplaza por valor numerico 0...n
```

```python
# alternativa más directa con loc
ventas.loc['Nov'] = [198,98,58,'No',1.98,0.4]
print(ventas)
```

Update (Actualizar celdas, columnas o nombres)

```python
print(ventas)
# Ene        41       17          66     No    1.43        0.1
# cambiar una celda cuyo dato es 41 y es cambiado a 141
print('Update..... enero')
# iloc 0 por enero y 0 por entradas (valor=41)
ventas.iloc[0,0] = 141 # lo mismo que loc
print(ventas)
# cambiar salidas en febrero de 54 a 154
print('Update..... febrero')
# ventas.loc[1, 'Salidas'] = 154 # tansforma de 154 a 154.0 por usar loc con indice numerico
# crea un nuevo atributo con valores de tipo float (para evitar perdida de datos)

# solucion loc con etiquetas
ventas.loc['Feb','Salidas'] = 154

print(ventas)
```

Crear un nuevo atributo

```python
# crear una nuevo atributo 'Precio' que resultara del
# producto entre la valoracion y el cambio
import numpy as np
ventas['Precio'] = np.abs(ventas['Valoracion'] * ventas['Cambio'])
print(ventas)
```

Modificación de todos los valores de un atributo dado

```python
# aumentar todos los precios un 10%
# df['Precio'] = df['Precio'] * 1.10  # Aumenta un 10% todos los precios
ventas['Precio'] = ventas['Precio'] * 1.10
print(ventas)
```

Renombrar un atributo y mantener sus datos sin cambios

```python
# renombrar un atributo
# • .rename(columns={'vieja': 'nueva'}): Cambia el 
# nombre de una o varias columnas 
# (o índices usando index={...}).python
ventas = ventas.rename(columns={'Precio': 'Precio_Pesos_AR'})
print(ventas)
```

Reemplazar todos los valores de un atributo dado que cumpla cierta condición

```python
# reemplazar el valor de toda el dataframe de un atributo en particular
# Usa el código con precaución.
# • .replace(viejo, nuevo): Busca y reemplaza valores específicos en todo el DataFrame o en una columna puntual.python
# df['Producto'] = df['Producto'].replace('Mouse', 'Mouse Gamer')
ventas['Entradas'] = ventas["Entradas"].replace(99, 9999)
print(ventas)
```

Elimina filas duplicadas completas o basadas en una columna específica con subset=['columna'].

```python
# df_sin_duplicados = df.drop_duplicates(subset=['Producto'])
import pandas as pd 

datos_filas = {
    'Ene': [150, True],
    'Feb': [230, True],
    'Mar': [180, False],
    'Abr': [180, False],
    'May': [180, False],
    'Jun': [256, True],
    'Jul': [115, False],
    'Ago': [125, False],
    'Sep': [195, False],
    'Oct': [200, False],
    'Nov': [180, False],
    'Dic': [90, True],
    'En': [150, True],
    'Ma': [180, False]
}
df_filas = pd.DataFrame.from_dict(datos_filas, orient='index', columns=['Ventas', 'Activo'], dtype=pd.Int64Dtype())
print(df_filas)

df_filas_sin_duplicado = df_filas.drop_duplicates(subset=['Ventas'])

print(df_filas_sin_duplicado)
```
