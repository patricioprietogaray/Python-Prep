# 💻 Enunciado del Ejercicio: 
Análisis de Carga y Normalización de Servidores
Imagina que eres un ingeniero de infraestructura y estás analizando el rendimiento de 50 servidores a lo largo de 24 horas. Los datos se han recolectado en una matriz bidimensional simulada de NumPy llamada matriz_servidores de forma que las filas representan a los servidores (del 0 al 49) y las columnas representan las horas del día (de la 0 a la 23). Cada valor en la matriz indica el uso de CPU en porcentaje (%).

# Tus objetivos son:

Generación de Datos:

Utiliza una semilla fija (np.random.seed(100)) para garantizar la reproducibilidad.

Crea la matriz matriz_servidores de forma aleatoria con valores de uso de CPU que oscilen entre el 10.0% y el 95.5%.

```python
import numpy as np 
```

```python
# semilla en 100 para garantizar la reproducibilidad
np.random.seed(100)
```

```python
minimo = 10.0
maximo = 95.5
filas = 50
columnas = 24
```

```python
matriz_servidores = np.random.uniform(low=minimo, high=maximo, size=(filas, columnas))
```

```python
matriz_servidores
```

Análisis Exploratorio por Columnas (Horas):

Calcula el promedio de uso de CPU por hora durante todo el día (deberá darte un array con 24 valores).

Identifica y muestra cuál fue la hora del día (el índice de la columna) en la que se registró el pico máximo promedio de uso de CPU en toda la granja de servidores.

```python
# SOLO UNA COLUMNA
# Analisis exploratorio por columnas (horas)
columna_hora_cero = matriz_servidores[:,0].mean()
columna_hora_cero
# columna_hora_cero
```

```python
# PROMEDIO DE CADA COLUMNA -> TODAS LAS COLUMNAS
# axis=0 -> columnas
promedio_columna_por_hora = np.mean(matriz_servidores, axis=0)
promedio_columna_por_hora
```

```python
# BUSCAR EL PICO MAXIMO DE UTILIZACION DE LA GRANJA DE SERVIDORES
hora_maxima_uso_servidores = np.max(promedio_columna_por_hora)
indice_maximo_uso = np.argmax(promedio_columna_por_hora)
print(f"La hora maxima de uso de los servidores es {indice_maximo_uso}hs con una carga de {hora_maxima_uso_servidores}!")

```

Análisis Exploratorio por Filas (Servidores):

Calcula el consumo máximo individual alcanzado por cada servidor a lo largo de las 24 horas.

Filtra y cuenta cuántos servidores superaron un uso pico de 90.0% de CPU en algún momento del día.

```python
import numpy as np 
consumo_maximo_servidor = np.max(matriz_servidores, axis=1)
hora_cons_max_serv = np.argmax(matriz_servidores, axis=1)
print(f"Todos los servidores")
for indice, consumo in enumerate(consumo_maximo_servidor): 
    print(f"Servidor nro {indice}, consumo maximo {consumo} a las {hora_cons_max_serv[indice]} horas!")
# print(consumo_maximo_servidor)
# print(hora_cons_max_serv)



```

```python
import numpy as np  
# servidores que superaron el 90.0%
consumo_mas_90 = consumo_maximo_servidor[consumo_maximo_servidor > 90.0]
consumo_mas_90
```

Normalización Min-Max (Preprocesamiento):
Aplica la fórmula de normalización Min-Max a toda la matriz para llevar 
los valores de uso de CPU a una escala entre 0.0 y 1.0:


                                       X - Xmin

                        X_norm = -------------------------


                                      Xmax - Xmin 
            

Comprueba que el valor mínimo global de tu matriz normalizada sea exactamente 0.0 y el máximo global sea 1.0 utilizando funciones de NumPy.

```python
X = matriz_servidores
# Xmin = 10.0
Xmin = X.min()
# Xmax = 95.5
Xmax = X.max()
X_norm = np.divide((X - Xmin),(Xmax - Xmin))
X_norm
```
