```python
# utilizar bucle for
# for elemento in elemento_a_iterar:
#       codigo_que_se_repetirá

# lista para iterar
numeros = [1,2,3,4,5]
for num in numeros:
    print(num+1)
```

```python
for i in range(1,6):
    # el ultimo valor de range no se incluye
    suma = i + 1
    print(suma)
```

```python
# resultado de la iteracion
for num in range(1,6):
    suma = num + 1
    print("Variable de la iteracion", num, ", Resultado: ", suma)
    
```

```python
# bucle while se ejecutará mientras la condicion
# sea verdadera

num=1
while num <6:
    print("num", num)
    num+=1
```

```python
# el bucle for se utiliza cuando se sabe de ante mano 
# la cantidad maxima de iteraciones
# ej: 
# voy a ver series hasta las 7 de la tarde


# el bucle while se utiliza cuando no se tiene la 
# certeza del numero de iteraciones que deba ejecutar
# ej: 
# mientras siga lloviendo voy a estar mirando series

# proximo flujos de control - 
#       sentencia break, continue
# break sale del bucle while o for
# continue continua al proximo ciclo del bucle

for i in range(5):
    print("Siempre entra: Valor de i: ", i)
    if i == 4:
        print("Salgo del bucle")
        break
    elif i == 1:
        print("Continuo en el bucle vuelve a la primer linea")
        continue
    print("ultima linea del bucle (i: ",i,")")
print("el bucle se terminó")
```

```python
cadena = "Python"
for letra in cadena:
    if letra == 'h':
        print("Se encontró la h")
        break
    print(letra)
print("Fin del bucle for")
```

```python
cadena = "Python"
contador=0


while contador < len(cadena):
    letra = cadena[contador]
    if letra == 'h':
        print("Se encontró la h")
        break
    print(letra)
    contador+=1
print("Fin del bucle while")
```
