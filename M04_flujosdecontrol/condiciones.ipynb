```python
a=5
b=7
if a<b:
    print("a es menor que b")
elif a>b:
    print("a es mayor que b")
else:
    print("a y b son iguales")
```

```python
# pregunta
# ejecuta solo la opcion correcta
# solo menor a 10 porque esta primera en 
# el orden secuencial 
# sale del condicional
# menor a 7 si bien es una opcion valida
# ejecuto menor a 19 y ya salió
# continua con la siguiente linea

num=5
if num > 10:
    print("Soy mayor!")
elif num < 10:
    print("soy menor a 10")
elif num < 7:
    print("soy menor a 7")
else:
    print("Soy igual")

print("fin del condicional if")
```
