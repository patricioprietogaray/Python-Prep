```python
# MANEJO DE ERRORES

# Es la capacidad de detectar y responder adecuadamente
# a los errores que pueden ocurrir durante la ejecución del código

# Algunos errores pueden ser: 
# entradas incorrectas por el ususrio, 
# recepcion de información desde un servidor, 
# problemas de conexion a una base de datos, 
# errores de sintaxis o lógica

# ¿Como verificar la calidad y funcionalidad del software?
# PRUEBAS DE CAJA NEGRA
# PRUEBAS DE CAJA DE CRISTAL 

# PRUEBA DE CAJA NEGRA
# Evaluar el funcionamiento de un programa sin necesidad 
# de conocer la estructura interna o lógica de funcionamiento.
# Se prueban los input y output del programa sin conocer 
# los procesos intermedios que suceden en esa parte del codigo. 


# Las pruebas de caja negra son un enfoque de pruebas en el que se evalúa 
# el funcionamiento de un sistema sin conocer su implementación interna. 
# En otras palabras, se prueba el sistema desde una perspectiva externa, 
# centrándose únicamente en las entradas y salidas esperadas, 
# sin tener conocimiento detallado de cómo se procesan los datos o 
# cómo se realizan las operaciones internas.

# Ejemplo: Prueba de caja negra seria el ingreso de información para un formulario web
# solo se verifica que la clave ingresada tenga ciertas características y 
# que el correo electrónico sea válido. 
# Como resultado es el registro del usuario; si el usuario ingresa los datos en forma correcta. 

```

```python
# PRUEBAS UNITARIAS 
# Se prueban las unidades de código
# es decir las porciones de código pequeñas y 
# porciones de código aisladas 

# No se tiene en cuenta su integración al sistema

# El objetivo es asegurarse que cada unidad de código 
# funcione por si sola, a un input determinado 
# mismo output determinado; que CUMPLA CON LAS ESPECIFICACIONES DEFINIDAS

# PRUEBAS DE INTEGRACION
# Se prueban varios componentes del software junto

# Simular el comportamiento del sistema completo

# El objetivo es verificar que los componentes individuales 
# funcionen correctamente cuando se integran con otros componentes

# Que el sistema completo funcione de acuerdo con las especificaciones

# Es conveniente que el sistema tenga ciertos tests para poder probar el código
# unitario, modular o total del sistema.


```

```python
# LIBRERIA UNITTEST

# Herramienta para realizar pruebas unitarias

# Evalua el comportamiento del codigo de forma aislada

# En Python existe la posibilidad de realizar estas pruebas gracias 
# a la librería unittest. 
# La librería unittest de Python es una herramienta 
# que se utiliza para realizar pruebas unitarias, 
# donde el enfoque está en evaluar el comportamiento de 
# unidades de código de forma aislada, como funciones o métodos, 
# para verificar que funcionan correctamente.


```

```python
import unittest
```

```python
def suma(num1, num2):
    return num1 + num2
```

```python
# crear un clase que hereda de unittest -> testcase

# en esta clase hay dos metodos de pruebas suma dos positivos y suma dos negativos
class CajaNegraTest(unittest.TestCase):
    
    def test_suma_dos_positivos(self):
        num_1 = 10
        num_2 = 5
        
        resultado = suma(num_1, num_2)
        
        # verifica resultado (el resultado es 15)
        self.assertEqual(resultado, 15)
        
    def test_suma_dos_negativos(self):
        num_1 = -7
        num_2 = -10
        
        resultado = suma(num_1, num_2)
        self.assertEqual(resultado, -17)
        
        
    
```

```python
# argv -> proporciona test para ejecutar una linea de comandos en la prueba
# exit -> las pruebas terminan luego de la ejecucion (True) o siguen indefinidamente (False)?
unittest.main(argv=[""], verbosity=2, exit=False)

# al utilizar el método __main__ se ejecutaran todos los metodos que se invoquen
# que se encuentran en la clase CajaNegraTest
test_suma_dos_positivos(__main__.CajaNegraTest)
test_suma_dos_negativos(__main__.CajaNegraTest)
```

```python
def es_mayor_de_edad(edad):
    if edad >= 18:
        return True
    else:
        return False
    
```

```python
class PruebaDeCristalTest(unittest.TestCase):
    def test_es_mayor_de_edad(self):
        edad = 20
        resultado = es_mayor_de_edad(edad)
        self.assertEqual(resultado, True)
    
    def test_es_menor_de_edad(self):
        edad = 10
        resultado = es_mayor_de_edad(edad)
        # self.assertEqual(resultado, False)
        
        # hago que salte un error en la prueba
        self.assertEqual(resultado, True)
```

```python
def divide_elementos_lista(lista, divisor):
    '''
    Cada elemento de una lista es dividida por el divisor definido.
    En caso de error de tipo ZeroDivisionError que significa
    error al dividir por cero 
    la funcion devuelve la lista inicial
    '''
    
    try:
        # comprension de lista
        return [i / divisor for i in lista]
    except ZeroDivisionError as e:
        print(e)
        return lista
```

```python
# Excepciones, control de flujo y afirmaciones

# En esta lección seguiremos profundizando sobre las excepciones 
# en Python, control de flujo y conoceremos las afirmaciones.

# En Python, las excepciones son eventos 
# que ocurren durante la ejecución de un programa y 
# que interrumpen su flujo normal. 
# Cuando una situación excepcional se produce, se lanza una excepción, 
# lo que permite al programador manejar y controlar el error de manera adecuada.


# Las excepciones pueden ser causadas por diferentes factores, 
# como errores de sintaxis, 
# intentos de acceder a elementos inexistentes, 
# divisiones entre cero o 
# cualquier otra condición imprevista. 
# Cuando se produce una excepción, el programa se detiene y 
# busca un bloque de código que pueda manejarla.

# try:
#     codigo a ejecutar
    
# except tipo_excepcion:
#     codigo que se ejecuta si se produce la excepcion 
    
```

```python
def division(n1,n2):
    try:
        resultado = n1 / n2
        return resultado
    except ZeroDivisionError as e:
        print(e)
        print("No se puede dividir por cero")
        return None
```

```python
resultado = division(8,0)
print(resultado)
```

```python
def primera_letra(lista_de_palabras):
    primeras_letras = []

    for palabra in lista_de_palabras:
        if not isinstance(palabra, str):
            raise TypeError(f'{palabra} no es un str')
        if len(palabra) == 0:
            raise ValueError('No se permiten str vacios')

        primeras_letras.append(palabra[0])

    return primeras_letras
```

```python
def primera_letra(lista_de_palabras):
    primeras_letras = []

    for palabra in lista_de_palabras:
        try:
            if not isinstance(palabra, str):
                raise TypeError(f'{palabra} no es un str')
            if len(palabra) == 0:
                raise ValueError('No se permiten str vacios')

            primeras_letras.append(palabra[0])
        except (TypeError, ValueError) as error:
            print("Error: ", error)

    return primeras_letras
```

```python
palabras = [12,'arg','',True]
print(primera_letra(palabras))
```
