```python
import unittest
```

```python
def suma(a, b):
    return a + b
```

```python
class CajaNegraTest(unittest.TestCase):
    def test_suma_dos_positivos(self):
        a = 10
        b = 5
        resultado = suma(a, b)
        self.assertEqual(resultado, 15)
    
    def test_suma_positivo_negativo(self):
        a = -10
        b = -5
        resultado = suma(a, b)
        self.assertEqual(resultado, -15)
```

```python
unittest.main(argv=[''], verbosity=2, exit=False)
# funcion_test (__main__.CajaNegraTest) ... ok

test_suma_dos_positivos(__main__.CajaNegraTest) 
test_suma_positivo_negativo(__main__.CajaNegraTest) 
```
