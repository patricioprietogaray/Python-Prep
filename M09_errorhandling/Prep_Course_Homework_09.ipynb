```python
# import unittest
from collections import Counter

class Numeros:
    def __init__(self, lista_numeros):
        # Validación: Comprueba que el argumento sea una lista
        if not isinstance(lista_numeros, list):
            raise TypeError("El argumento debe ser una lista.")
        # Validación: Comprueba que la lista no esté vacía
        if not lista_numeros:
            raise ValueError("La lista no puede estar vacía.")
        # Validación: Comprueba que todos los elementos sean números 
        # (int o float)
        if not all(isinstance(num, (int, float)) for num in lista_numeros):
            raise TypeError("Todos los elementos de la lista deben ser números (int o float).")
        
        self.lista_numeros = lista_numeros
        self.numero_primo = lista_numeros[0]

    def es_primo(self):
        if self.numero_primo < 2:
            return False
        for i in range(2, int(self.numero_primo ** 0.5) + 1):
            if self.numero_primo % i == 0:
                return False
        return True
    
    def convertir_temperatura(self, valor, origen, destino):
        origen = origen.upper()
        destino = destino.upper()
            
        if origen == 'C' and destino == 'F':
            return (float(valor) * (9/5)) + 32
        elif origen == 'C' and destino == 'K':
            return float(valor) + 273.15
        elif origen == 'F' and destino == 'C':
            return (float(valor) - 32) * (5/9)
        elif origen == 'F' and destino == 'K':
            return ((float(valor) - 32) * (5/9)) + 273.15
        elif origen == 'K' and destino == 'C':
            return float(valor) - 273.15
        elif origen == 'K' and destino == 'F':
            return ((float(valor) - 273.15) * (9/5)) + 32
    
    def valor_modal(self):
        """Calcula la moda (valor con mayor frecuencia) de la lista de números."""
        conteo = Counter(self.lista_numeros)
        max_frecuencia = max(conteo.values())
        # return (conteo, max_frecuencia)
        modas = [num for num, freq in conteo.items() if freq == max_frecuencia]
        # Devuelve el valor si es único, o la lista si hay múltiples modas con la misma frecuencia
        return modas[0] if len(modas) == 1 else modas


    def verifica_primos_lista(self):
        """Devuelve una lista de números primos encontrados en la lista de números."""
        lista_primos = []
        for num in self.lista_numeros:
            self.numero_primo = num  # Actualiza el número a verificar
            if self.es_primo():
                lista_primos.append(True)
            else:
                lista_primos.append(False)
        return lista_primos

    def factorial(self, n):
        """Calcula el factorial de un número entero no negativo."""
        if not isinstance(n, int) or n < 0:
            raise ValueError("El número debe ser un entero no negativo.")
        if n == 0 or n == 1:
            return 1
        resultado = 1
        for i in range(2, n + 1):
            resultado *= i
        return resultado

# Utilizar raise ValueError() dentro de __init__ 
# es una práctica fundamental de diseño 
# defensivo y programación orientada a objetos 
# por las siguientes razones:

# Fail-fast (Falla rápido): 
# Permite detectar estados de datos inválidos 
# en el preciso momento en que se intenta 
# crear el objeto, evitando que la aplicación 
# continúe ejecutándose con datos corruptos o inconsistentes.

# Integridad del Estado: 
# Impide que se cree un objeto "roto" o incompleto en la memoria.

# Pruebas unitarias sencillas: 
# En unittest, se puede verificar 
# fácilmente este comportamiento utilizando 
# el gestor de contexto 
# with self.assertRaises(ValueError):, 
# el cual comprueba que efectivamente se 
# lance la excepción esperada cuando los datos de 
# entrada son incorrectos.

# Código Completo con las Nuevas Pruebas 
# y el Método valor_modal()
# A continuación se presenta la clase 
# actualizada que incluye la validación 
# en el constructor con raise ValueError(), 
# el nuevo método valor_modal() 
# (utilizando la clase Counter para calcular 
# la moda estadística) y la suite de pruebas 
# con unittest que evalúa los tres escenarios solicitados.
```

```python
import unittest

class TestNumeros(unittest.TestCase):
    def setUp(self):
        # Se crea una instancia reutilizable para los tests
        self.temp_obj = Numeros([0, 10, 20])

    def test_creacion_objeto_incorrecta(self):
        with self.assertRaises(TypeError):
            Numeros("12345")  # No es una lista
        
        with self.assertRaises(ValueError):
            Numeros([])  # Lista vacía
        
    def test_es_primo(self):
        primo = Numeros([170])
        # devuelve False porque 170 no es primo
        self.assertFalse(primo.es_primo())
        
        # devuelve True porque 17 es primo
        self.assertTrue(Numeros([17]).es_primo())

        primo = Numeros([11])
        self.assertTrue(primo.es_primo())
        # devuelve True porque 11 es primo


    def test_convertir_temperatura(self):
        resultado = self.temp_obj.convertir_temperatura(10.0, 'C', 'F')
        self.assertEqual(resultado, 50.0)

        resultado = self.temp_obj.convertir_temperatura(0, 'F', 'C')
        self.assertEqual(resultado, -17.77777777777778)

        resultado = self.temp_obj.convertir_temperatura(273.15, 'K', 'C')
        self.assertEqual(resultado, 0.0)

    def test_valor_modal(self):
        obj_moda = Numeros([0, 10, 20, 10, 30, 10])
        self.assertEqual(obj_moda.valor_modal(), 10)
        
```
