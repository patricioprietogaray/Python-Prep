```python
# organizar el testing de una manera particular y mas
# escalable

def sumar(a, b):
    return a + b

def restar(a, b):
    return a - b

def multiplicar(a, b):
    return a * b

def dividir(a, b):
    if b == 0:
        raise ValueError("No se puede dividir por cero.")
    return a / b

```

```python
import unittest

class TestOperaciones(unittest.TestCase):
    def test_sumar(self):
        self.assertEqual(sumar(2, 3), 5)
        self.assertEqual(sumar(-1, 1), 0)

    def test_restar(self):
        self.assertEqual(restar(5, 3), 2)
        self.assertEqual(restar(0, 5), -5)

    def test_multiplicar(self):
        self.assertEqual(multiplicar(2, 3), 6)
        self.assertEqual(multiplicar(4, 3), 12)
        self.assertEqual(multiplicar(-2, 3), -6)

    def test_dividir(self):
        self.assertEqual(dividir(10, 2), 5)
        with self.assertRaises(ValueError):
            dividir(10, 0)
```

```python
class Numeros:
    def __init__(self, lista_numeros):
        if not isinstance(lista_numeros, list):
            raise TypeError("El argumento debe ser una lista.")
        if not lista_numeros:
            raise ValueError("La lista no puede estar vacía.")
        for numero in lista_numeros:
            if not isinstance(numero, (int, float)):
                raise TypeError("Todos los elementos de la lista deben ser números (int o float).")
    

# objeto = Numeros([1, 2, '3', 4.5, 5.6])
```

```python
# restricciones 
```

```python
# clase test

class TestNumeros(unittest.TestCase):
    def test_tipo_argumento(self):
        with self.assertRaises(TypeError):
            Numeros("no es una lista")

    def test_lista_vacia(self):
        with self.assertRaises(ValueError):
            Numeros([])

    def test_elementos_no_numericos(self):
        with self.assertRaises(TypeError):
            Numeros([1, 2, '3', 4.5, 5.6])
```
