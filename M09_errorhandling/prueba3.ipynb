```python
# cargar todas las librerias necesarias
import unittest
```

```python
# crear la clase pura 

class Numeros_Primos:
    def __init__(self, n):
        self.n = n

    def es_primo(self):
        if self.n < 2:
            return False
        for i in range(2, int(self.n ** 0.5) + 1):
            if self.n % i == 0:
                return False
        return True


```

```python
# crear las restricciones de la clase pura

class Restricciones_Numeros_Primos(Numeros_Primos):
    def __init__(self, n):
        super().__init__(n)

    def es_primo(self):
        if not isinstance(self.n, int):
            raise ValueError("El número debe ser un entero.")
        if self.n < 0:
            raise ValueError("El número debe ser positivo.")
        return super().es_primo()
```

```python
# ejecutar la clase pura con las restricciones
restriccion = Restricciones_Numeros_Primos(17)
print(restriccion.es_primo())
```

```python
# ejecutar la clase pura con las restricciones
restriccion = Restricciones_Numeros_Primos(-5)
print(restriccion.es_primo())
```

```python
# crear los test de la clase pura
class TestNumerosPrimos(unittest.TestCase):
    def test_es_primo(self):
        self.assertTrue(Restricciones_Numeros_Primos(2).es_primo())
        self.assertTrue(Restricciones_Numeros_Primos(3).es_primo())
        self.assertFalse(Restricciones_Numeros_Primos(4).es_primo())
        self.assertTrue(Restricciones_Numeros_Primos(5).es_primo())
        self.assertFalse(Restricciones_Numeros_Primos(6).es_primo())
        self.assertTrue(Restricciones_Numeros_Primos(7).es_primo())
        self.assertFalse(Restricciones_Numeros_Primos(8).es_primo())
        self.assertFalse(Restricciones_Numeros_Primos(9).es_primo())
        with self.assertRaises(ValueError):
            Restricciones_Numeros_Primos(-1).es_primo()
            Restricciones_Numeros_Primos(-2).es_primo()
        # self.assertFalse(Restricciones_Numeros_Primos(-1).es_primo())
        # self.assertFalse(Restricciones_Numeros_Primos(-2).es_primo())
```

```python
# correr los test

unittest.main(argv=[''], verbosity=2, exit=False)
# test_es_primo(__main__.TestNumerosPrimos)

```
