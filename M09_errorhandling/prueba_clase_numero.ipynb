```python
class Numeros:
    def __init__(self, nros=[]):
        self.nros = nros
    
    def verificar(self):
        if len(self.nros) == 0:
            raise ValueError("La lista de números está vacía.")
        else:
            return f"La lista contiene {len(self.nros)} números."    
```

```python
import unittest

class NumerosTest(unittest.TestCase):
    def test_verificar_lista_vacia(self):
        nros = Numeros()
        with self.assertRaises(ValueError):
            nros.verificar()

    def test_verificar_lista_con_elementos(self):
        nros = Numeros([1, 2, 3])
        self.assertEqual(nros.verificar(), "La lista contiene 3 números.")
```

```python
unittest.main(argv=[''], verbosity=2, exit=False)
```
