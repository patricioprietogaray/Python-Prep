```python
lista_num_des = [5,6,1,8,2,9,4,3]
print(f"original {lista_num_des}")
# apuntan al mismo lugar de la memoria (como el puntero en c)
# lista_num_ord_asc = lista_num_des
lista_num_ord_asc = lista_num_des.copy()
print(f"copia del original para ordenar: {lista_num_des}")
lista_num_ord_asc.sort()
print(f"lista ordenada ascendentememnte: {lista_num_ord_asc}")
lista_num_ord_desc = lista_num_des.copy()
print(f"copia del original para ordenar: {lista_num_des}")
lista_num_ord_desc.sort(reverse=True)
print(f"lista ordenada desendentememnte: {lista_num_ord_desc}")
```
