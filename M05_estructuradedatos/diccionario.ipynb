```python
# diccionarios
# son mutables
# secuencia de valores con indices de cualquier tipo
# claves y valores
# las claves apuntan a un valor
# los diccionarios van entre llaves {}
# para acceder a un valor solo es por su clave
# no se acceden a un valor por su indice
# para crear un diccionario usar Dict
primer_diccionario = dict()
print(primer_diccionario)

```

```python
#agregar un elemento
primer_diccionario['primero']='uno'
primer_diccionario['segundo']='dos'
primer_diccionario['tercero']='tres'
print(primer_diccionario)
```

```python
segundo_diccionario={'cuatro':4, 'cinco':5, 'seis':6}
print(segundo_diccionario)
```

```python
# acceder a un valor especifico del diccionario
print(primer_diccionario['primero'])
print(segundo_diccionario['cinco'])
```

```python
mi_diccionario = {
    'Colores Primarios':['Rojo','Azul','Amarillo'],
    'Colores Secundarios': ['Naranja','Violeta','Verde'],
    'Clave 3': 10,
    'Clave 4': False
}
```

```python
print(mi_diccionario['Clave 4'])
```

```python
# borrar una clave valor del diccionario con del
del mi_diccionario['Clave 3']
print(mi_diccionario)
```

```python
# asignar datos en un diccionario con clave desde elementos
# desde una tupla

mi_tupla = ('Argentina','Italia','Inglaterra')
mi_diccionario={
    mi_tupla[0]:'Buenos Aires',
    mi_tupla[1]:'Roma',
    mi_tupla[2]:'Londres'
}
print(mi_diccionario)
```

```python
# colocar una tupla dentro de un diccionario
mi_diccionario = {
    'Clave1':'Valor 1',
    'Clave2': (1,2,3,4,5)
}
```

```python
# muestro el valor buscando por clave
print(mi_diccionario['Clave1'])

# muestro toda la tupla
print(mi_diccionario['Clave2'])

# muestro el valor y accedo al indice de la tupla
print(mi_diccionario['Clave2'][0])
```

```python
# colocar una lista dentro de un diccionario
mi_diccionario = {
    'Clave1':'Valor 1',
    'Clave2': [1,2,3,4,5]
}
```

```python
# muestro el valor buscando por clave
print(mi_diccionario['Clave1'])

# muestro toda la lista
print(mi_diccionario['Clave2'])

# muestro el valor y accedo al indice de la lista
print(mi_diccionario['Clave2'][0])
```

```python
# colocar una lista dentro de un diccionario
mi_diccionario = {
    'Clave1':'Valor 1',
    'Clave2': {'uno':1,'dos':2,'tres':3}
}
print(mi_diccionario)
print(mi_diccionario['Clave2']['uno'])
```

```python
# para obtener las claves del diccionario
print(mi_diccionario.keys())
```

```python
# para obtener los valores del diccionario
print(mi_diccionario.values())
```

```python
# para saber la cantidad de 
# pares clave-valor de un diccionario
print(len(mi_diccionario))
# 10.59
```
