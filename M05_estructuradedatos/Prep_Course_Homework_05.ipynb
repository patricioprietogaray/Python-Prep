23) Crear un diccionario utilizando la lista crada en el punto 1, asignandole la clave "ciudad". Agregar tambien otras claves, como puede ser "Pais" y "Continente".

```python
ciudades = ["Buenos Aires", "Bogotá", "Londres", "Tokio", "Barcelona", "Pekin", "Otawa", "Camberra"]
pais = ['Argentina','Colombia','Inglaterra','Japón','España', 'China', 'Canadá', 'Australia']
continente = ('America','América','Europa','Asia','Europa','Asia','América','Oceanía')
lista_diccionario = list()
for i in range(0, len(ciudades)):
    datos_ciudad = {
        'ciudad': ciudades[i],
        'pais': pais[i],
        'continente': continente[i]
    }
    lista_diccionario.append(datos_ciudad)
print(lista_diccionario)
```

24) Imprimir las claves del diccionario

```python
print(lista_diccionario)
print(lista_diccionario[0].keys())
```

25) Imprimir las ciudades a través de su clave

```python
for i in range(0, len(lista_diccionario)):
    print(lista_diccionario[i]['ciudad'])
```
