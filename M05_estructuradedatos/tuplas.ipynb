```python
# empaquetado de tuplas, de una tupla asigno varia variables
# a fin de cargar sus datos
# Podemos generar una tupla asignando a una variable un conjunto de variables
# separadas por coma, a esto se le llama empaquetado de tuplas.

x
```

```python
x = 11
y = 22
z = "Henry"
# se asignan los valores de las variables a la tupla (siempre se almacenara en una tupla)
mi_tupla_prueba = x,y,z
print(mi_tupla_prueba)
print(type(mi_tupla_prueba))
```

```python
tupla_fecha = (22,"agosto", 1999)
print(tupla_fecha)
print(type(tupla_fecha))
```

```python
# desempaquetar una tupla
dia,mes,anio = tupla_fecha
print(dia)
print(mes)
print(anio)
```

```python
# uno,dos=tupla_fecha
# ---------------------------------------------------------------------------
# ValueError                                Traceback (most recent call last)
# Cell In[29], line 1
# ----> 1 uno,dos=tupla_fecha

# ValueError: too many values to unpack (expected 2, got 3)
# demasiados valores para desempaquetar (se esperan 3 variables declaradas y no dos variables)
# lo mismo pasa si hay mas variables que valores disponibles
```

```python
tupla_impares = (1,3,5,7,9)
primero = tupla_impares[0]
cuarto = tupla_impares[3]
quinto = tupla_impares[4]
print(primero)
print(cuarto)
print(quinto)
```

```python
a,b,c,d,e = tupla_impares
print(f"{a}, {b}, {c}, {d}, {e}")
# solo desempaquetar algunos elementos dd y ee
_,_,_,dd,ee = tupla_impares
print(f"{dd}, {ee}")
# print(_) -> muestra 5 porque 1 es reemplazado por 3 y luego por 5
```

```python
lista_impares = list(tupla_impares)
print(lista_impares)
print(type(lista_impares))
```

```python
tupla_para_impares = tuple(lista_impares)
print(tupla_para_impares)
print(type(tupla_para_impares))
```
