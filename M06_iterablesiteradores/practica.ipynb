```python
# iteracion con clase_iterable -> diccionario
mi_dict = {'a':1, 'b':2, 'c':3}
for i in mi_dict:
    print(f'Imprimir la clave: {i}')
    print(f'Imprimir su valor {mi_dict[i]}')
    print("+"*15)
```

```python
# enumerate(mi_dict) devuelve en cada iteración 
# una tupla de 2 elementos: 
# (índice, clave). -> [for i,c in enumerate(mi_dict)]
# 
# En este caso mi_dict se recorre como diccionario, 
# y su valor iterado es la clave 'a', 'b', 'c'.

# Por eso el for intenta asignar 
# 3 variables -> [for i,k,v in enumerate(mi_dict)]
# i, k, v a una tupla de solo 2 valores, 
# y Python lanza ValueError: not enough values to unpack 
# (expected 3, got 2).
# para solucionarlo k y v se ponen como una tupla


mi_dict = {'a':1, 'b':2, 'c':3}
for i,(k,v) in enumerate(mi_dict.items()):
    print(f'Indice: {i}')
    print(f'Imprimir la clave: {k}')
    print(f'Imprimir su valor {v}')
    print("+"*15)
```

```python
lista=[0,1,2,3]
marcador = iter(lista)
for c in lista:
    print(c)
    next(marcador)
# El fragmento crea una lista lista y un iterador explícito marcador = iter(lista). El bucle for recorre la lista e imprime cada elemento c. Dentro del cuerpo del bucle se llama a next(marcador), que avanza el iterador externo pero descarta su valor.

# Importante: el for crea y usa su propio iterador interno sobre lista, independiente de marcador. Por eso el código imprime todos los elementos (0, 1, 2, 3) mientras que marcador también avanza en paralelo; las llamadas a next(marcador) aquí son redundantes y podrían provocar una excepción StopIteration si se llamaran más veces de las que hay elementos.
```

```python
lista=[100,101,102,103,104] 
marcador = iter(lista)
while True:
    try:
        print(next(marcador))
    except StopIteration:
        print("Ya se paso del ultimo elemento, no hay nada mas que iterar")
        break
```

```python
# añadiendo condicionales

# En la programación, 
# las condiciones son un aspecto fundamental para controlar 
# el flujo de nuestro código. 
# En Python, podemos utilizar declaraciones 
# condicionales como 'if', 'else' y 'elif' para 
# tomar decisiones basadas en ciertas condiciones. 
# En esta clase, aprenderemos cómo utilizar estas 
# estructuras de control de flujo para agregar 
# condiciones a nuestro código, 
# lo que nos permitirá crear programas más complejos 
# y adaptables a distintas situaciones. ¡Vamos a empezar!

# ATENCION
# lista = [expresion for elemento in iterable if condicion]
# LISTA -> se creará una nueva LISTA a partir de la comprensión
#             de listas. 
# EXPRESION -> la operación que se realizará en cada elemento
#                 del iterable para crear el elemento de la 
#                 nueva lista. (expresiones matemáticas, llamada 
#                 a una funcion, concatenacion, etc )
# ELEMENTO -> es la variable que representa a cada elemento del 
#                 iterable original. 
# ITERABLE -> la secuencia original de la cual se creará la nueva 
#             lista. (lista, tupla, rango, conjuntos, etc) 
# IF -> expresión booleana que se evalua a cada uno de los elementos 
#         del iterable original. 
#         si no hay expresion (if condicion) se aplicará la expresion 
#         a todos los elementos del iterable. 

# ejemplo de comprensión de clase iterable
lista = [1,2,3,4,5,6]
# proceso de creacion de la instruccion

# pares_por_dos = [for elemento in lista] 
# #recorro la lista
# sin salida, error por falta de parámetros

# pares_por_dos = [elemento*2 for elemento in lista] 
# #agrego la salida antes del recorrido
# salida: "Recorro la lista: [2, 4, 6, 8, 10, 12]"

# agrego al final el condicional que se ejecutará siempre y 
# en cada elemento recorrido por el bucle for
pares_por_dos = [elemento*2 for elemento in lista if elemento%2==0]
print(f"Recorro la lista: {pares_por_dos} ")
```
